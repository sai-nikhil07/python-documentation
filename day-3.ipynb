# Conditional Control Statements in Python


In Python, If-Else is a fundamental conditional statement used for decision-making in programming. It allows your code to dynamically respond to different inputs and situations.

## 1. Definition

An **If...Else statement** allows the execution of specific blocks of code depending on whether the condition is `True` or `False`. They alter the sequential flow of execution, enabling the program to make choices.

## 2. Real-World Examples

- **E-commerce Checkout:** **If** the user's cart total is greater than $50, apply free shipping. **Else**, add a $5 shipping fee.
- **Authentication System:** **If** the entered password matches the database, log the user in. **Else**, display an "Invalid password" error.
- **Smart Thermostat:** **If** the room temperature falls below 68Â°F, turn on the heater. **Elif** it goes above 78Â°F, turn on the AC.

## 3. Applications

- **Form Validation:** Checking if user inputs meet specific criteria (e.g., email format is valid, age > 18).
- **Game Logic:** Determining outcomes based on player actions (e.g., if player health <= 0, trigger "Game Over" screen).
- **Data Filtering:** Processing datasets and extracting only the records that meet certain conditions.
- **Routing in Web Apps:** Showing different web pages based on the URL the user requested.

## 4. Syntax

We can break down Python conditionals into the following main categories:

### A. `if` Statement
The `if` statement is the most simple decision-making statement. If the condition evaluates to `True`, the block of code inside the `if` statement is executed.

```python
i = 10
# Checking if i is greater than 15
if i > 15:
    print("10 is less than 15")
print("I am Not in if")
```

### B. `if....else` Statement
When the `if` condition is `False`, the `else` block will be executed.

```python
i = 20
# Checking if i is greater than 0
if i > 0:
    print("i is positive")
else:
    print("i is 0 or Negative")
```

### C. If Else in One-line (Shorthand / Ternary)
If we need to execute a single statement inside the `if` or `else` block, a one-line shorthand can be used.

```python
a = -2
# Ternary conditional to check if number is positive or negative
res = "Positive" if a >= 0 else "Negative"
print(res)
```

### D. Logical Operators with If..Else
We can combine multiple conditions using logical operators such as `and`, `or`, and `not`.

```python
age = 25
exp = 10
# Using '>' operator & 'and' with if-else
if age > 23 and exp > 8:
    print("Eligible.")
else:
    print("Not eligible.")
```

### E. Nested `if...else` Statement
This occurs when an `if...else` structure is placed inside another `if` or `else` block. It allows conditional checks based on a series of conditions.

```python
i = 10
if i == 10:
    # First if statement
    if i < 15:
        print("i is smaller than 15")
        # Nested if statement
        if i < 12:
            print("i is smaller than 12 too")
        else:
            print("i is greater than 15")
else:
    print("i is not equal to 10")
```

### F. `if-elif-else` Statement (Ladder)
Used for multi-way decision-making. Checks multiple conditions sequentially.

```python
i = 25
if i == 10:
    print("i is 10")
elif i == 15:
    print("i is 15")
elif i == 20:
    print("i is 20")
else:
    print("i is not present")
```

### G. `Match Case ` 
match-case statement is Python's version of a switch-case found in other languages. It allows us to match a variable's value against a set of patterns.

```python
number = 2

match number:
    case 1:
        print("One")
    case 2 | 3:
        print("Two or Three")
    case _:
        print("Other number")
```

## 5. Rules and Characteristics

- **Boolean Evaluation:** The condition must evaluate to a boolean value (`True` or `False`). Python also considers values like `0`, `None`, empty lists `[]`, and empty strings `""` as `False` (falsy values).
- **Indentation is Mandatory:** The code block inside a conditional statement must be indented (usually 4 spaces). This is how Python determines the scope of the block.
- **Optional Clauses:** The `elif` and `else` blocks are optional.
- **Multiple Elifs:** You can string together as many `elif` conditions as needed, but you can only have a maximum of one `else` block at the very end.

## 6. Example with a LeetCode Problem

**Problem: 412. Fizz Buzz**

Given an integer `n`, return a string array `answer` (1-indexed) where:
- `answer[i] == "FizzBuzz"` if `i` is divisible by 3 and 5.
- `answer[i] == "Fizz"` if `i` is divisible by 3.
- `answer[i] == "Buzz"` if `i` is divisible by 5.
- `answer[i] == i` (as a string) if none of the above conditions are true.

```python
def fizzBuzz(n: int):
    answer = []
    for i in range(1, n + 1):
        # The if-elif-else ladder ensures only one block executes per iteration
        if i % 3 == 0 and i % 5 == 0:
            answer.append("FizzBuzz")
        elif i % 3 == 0:
            answer.append("Fizz")
        elif i % 5 == 0:
            answer.append("Buzz")
        else:
            answer.append(str(i))
    return answer

print(fizzBuzz(15))
```

## 7. Advantages and Disadvantages

### Advantages
- **Dynamic Execution:** Makes programs responsive to different inputs and states.
- **Core Problem Solving:** Essential for implementing algorithms and complex business logic.
- **Readability:** Clear `if-elif-else` structures explicitly show the decision-making process.

### Disadvantages
- **Spaghetti Code:** Deeply nested `if-else` statements make code incredibly difficult to read and maintain.
- **Cyclomatic Complexity:** Excessive conditionals increase the number of possible execution paths, making it harder to test.
- **Maintenance Overhead:** If rules change often, a long chain of `if-elif` statements can become brittle.

# Looping Control Statements in Python


In Python, loops are used to iterate over a sequence (like a list, tuple, dictionary, set, or string) or execute a block of code repeatedly as long as a condition is met.

## 1. Definition

 **Loops** in Python allow us to execute a group of statements multiple times. This is essential for automating repetitive tasks and efficiently processing collections of data.

## 2. Real-World Examples

- **Data Processing:** Iterating through millions of user records in a database to calculate average age.
- **Game Loop:** A `while True` loop that keeps a video game running, constantly listening for user input and updating the screen until the player quits.
- **File Reading:** Reading a text file line-by-line until the end of the file is reached.

## 3. Applications

- **Automating Repetitive Tasks:** Sending a welcome email to a list of 10,000 new subscribers.
- **Web Scraping:** Navigating through multiple pages of a website and extracting specific data from each.
- **Machine Learning:** Training a model by passing data through the neural network over multiple *epochs* (iterations).
- **Searching and Sorting Algorithms:** Implementing algorithms like Bubble Sort or Binary Search.

## 4. Syntax 

highlights two main types of loops in Python: `while` loops and `for` loops, along with loop control statements.

### A. `while` Loop
A `while` loop is used to execute a block of statements repeatedly until a given condition is satisfied. And when the condition becomes false, the line immediately after the loop in the program is executed.

```python
count = 0
while (count < 3):
    count = count + 1
    print("Hello World")
```

### B. `for` Loop
The `for` loop in Python is used for sequential traversal, i.e., it is used for iterating over an iterable like String, Tuple, List, Set, or Dictionary.

```python
# Iterating over a list
l = ["apple", "banana", "cherry"]
for i in l:
    print(i)
```

### C. Nested Loops
Python programming language allows using one loop inside another loop. This is known as nesting.

```python
for i in range(1, 4):     # Outer loop
    for j in range(1, 4): # Inner loop
        print(i, j)
```

### D. Loop Control Statements
Loop control statements change execution from its normal sequence. Python supports `break`, `continue`, and `pass` statements.

```python
# break: brings control out of the loop
for letter in 'pythonprogramming':
    if letter == 'e' or letter == 's':
        break
print('Current Letter (break):', letter)

# continue: returns control to the beginning of the loop
for letter in 'python':
    if letter == 'e' or letter == 's':
        continue
    print('Current Letter (continue):', letter)

# pass: writes empty loops (null statement)
for letter in 'python':
    pass
print('Last Letter:', letter)
```

## 5. Rules and Characteristics

- **`for` vs `while`:** Use `for` loops when you know the number of iterations in advance (like length of a list). Use `while` loops when the number of iterations is unknown and depends on a dynamic condition.
- **Infinite Loops:** If a `while` loop's condition never becomes `False`, it will run forever. Always ensure the loop has a way to terminate (e.g., incrementing a counter).
- **`else` with loops:** Python allows an optional `else` block with loops. The `else` block executes *only* if the loop terminates normally (i.e., not interrupted by a `break` statement).

## 6. Example with a LeetCode Problem

**Problem: 344. Reverse String**

Write a function that reverses a string. The input string is given as an array of characters `s`.
You must do this by modifying the input array in-place with O(1) extra memory.

```python
def reverseString(s: list) -> None:
    """
    Do not return anything, modify s in-place instead.
    """
    # We use a while loop with two pointers (start and end)
    left = 0
    right = len(s) - 1
    
    while left < right:
        # Swap the characters
        s[left], s[right] = s[right], s[left]
        # Move the pointers towards the center
        left += 1
        right -= 1

# Test the function
string_array = ["h","e","l","l","o"]
reverseString(string_array)
print(string_array) # Output: ['o', 'l', 'l', 'e', 'h']
```

## 7. Advantages and Disadvantages

### Advantages
- **Code Reusability:** Prevents writing the same code multiple times, adhering to the DRY (Don't Repeat Yourself) principle.
- **Efficiency:** Allows processing large volumes of data (like traversing arrays or matrices) with very few lines of code.
- **Flexibility:** With control statements like `break` and `continue`, loops can be highly customized to handle complex iteration logic.

### Disadvantages
- **Infinite Loops:** A poorly constructed `while` loop can crash a program or consume all system resources by running infinitely.
- **Performance Overhead:** In deeply nested loops (e.g., $O(n^2)$ or $O(n^3)$ time complexity), processing large datasets can become incredibly slow. Vectorization (like in NumPy) is often preferred over raw Python loops for heavy data crunching.
- **Off-by-One Errors:** It is common for beginners to make logic errors with loop boundaries (e.g., looping one time too many or one time too few).
