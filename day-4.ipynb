#### 1. Sum of digits
Problem Statement:
Add all digits of a number and return the total.

Input: 123

Output: 6

Explanation: 1 + 2 + 3 = 6

```python
n= 1234
s=0
while n:
    s+=n%10
    n=n//10
print(s)
```

#### 3. Palindrome Number
Problem Statement:

Check if number reads same forward and backward.

Test Case 1 :

Input: 121

```python
n=212
t=n
rev=0
while n:
    rem=n%10
    rev=rev*10+rem
    n//=10
print(f"value before : {t}, value after reverse :{rev}")

print("palindrome") if t==rev else print("not palindrome")
```

#### 8. Strong Number
Problem Statement :
Sum of factorial of digits equals number.

Test Case 1 :

Input: 145

Output: Strong

```python
n=145
t=n
su=0

while n :
    rem=n%10
    pr=1
    for i in range(1,rem+1):
        pr*=i
    su+=pr
    n//=10
print("Strong" if t==su else "not strong")
```

#### 20. Prime Digits Only
Problem Statement :
Check whether all digits of the number are prime digits.

Input:
num : 2357

Output:
All Prime Digits

```python
n=2357
t=n

while t:
    last=t%10
    # print(last)
    if last < 2:
        print("not all primes")
        break

    for i in range (2,int(last**0.5) + 1):
        if last%i==0:
            print("not all primes")
            break
    else:
        t //= 10
        continue
        
    break
else : 
    print("all primes")
```

#### 26. Decimal to Binary

#### 27. Binary to Decimal
Problem Statement :
Convert a binary number (containing only 0s and 1s) into its decimal equivalent.

Input:
num : 1010

Output:
10


```python
n=1010
de=0
p=0
while n:
    rem=n%10
    if rem >1:
        print("not a binary")
        break
    de=de+(2**p)*rem
    p+=1
    n//=10
else: print(de)
```

#### 39. Magic Number
Problem Statement : A number is called a Magic Number if the repeated sum of its digits (digital root) equals 1.

Input:
num : 19

Output:
Magic

```python
n= 19
while n > 9:
    s=0
    while n:
        rem=n%10
        s=s+rem
        n//=10
    n=s

print("magic number " if n==1 else "not magic ")
```
