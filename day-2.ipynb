## Day 2: Operators in Python
Operators are special symbols in Python that carry out arithmetic or logical computation. The value that the operator operates on is called the operand.

### 1. Arithmetic Operators
Arithmetic operators are used to perform mathematical operations like addition, subtraction, multiplication, etc.
- `+` Add
- `-` Subtract
- `*` Multiply
- `/` Divide
- `//` Floor Division
- `%` Modulo
- `**` Power

````python
a = 10
b = 3
print('Addition:', a + b)
print('Subtraction:', a - b)
print('Multiplication:', a * b)
print('Division:', a / b)
print('Floor Division:', a // b)
print('Modulo:', a % b)
print('Power:', a ** b)
````

### 2. Comparison (Relational) Operators
Comparison operators are used to compare values. It returns either `True` or `False` according to the condition.
- `==` Equal to
- `!=` Not equal to
- `>` Greater than
- `<` Less than
- `>=` Greater than or equal to
- `<=` Less than or equal to

````python
x = 10
y = 12
print('x > y is', x > y)
print('x < y is', x < y)
print('x == y is', x == y)
print('x != y is', x != y)
print('x >= y is', x >= y)
print('x <= y is', x <= y)
````

### 3. Logical Operators
Logical operators are used to combine conditional statements.
- `and` Returns True if both statements are true
- `or` Returns True if one of the statements is true
- `not` Reverse the result, returns False if the result is true

````python
a = True
b = False
print('a and b is', a and b)
print('a or b is', a or b)
print('not a is', not a)
````

### 4. Assignment Operators
Assignment operators are used to assign values to variables.
- `=` Assign
- `+=` Add AND
- `-=` Subtract AND
- `*=` Multiply AND
- `/=` Divide AND
- `//=` Floor Divide AND
- `%=` Modulus AND
- `**=` Exponent AND

````python
a = 10
a += 5 # Equivalent to a = a + 5
print('a += 5:', a)
a *= 2 # Equivalent to a = a * 2
print('a *= 2:', a)
````

### 5. Identity Operators
Identity operators are used to compare the memory locations of two objects.
- `is` Returns True if both variables are the same object
- `is not` Returns True if both variables are not the same object

````python
x1 = 5
y1 = 5
x2 = 'Hello'
y2 = 'Hello'
x3 = [1,2,3]
y3 = [1,2,3]

print('x1 is y1:', x1 is y1)
print('x2 is y2:', x2 is y2)
print('x3 is y3:', x3 is y3) # False because lists are mutable and have different memory locations
````

### 6. Membership Operators
Membership operators are used to test whether a value or variable is found in a sequence (string, list, tuple, set and dictionary).
- `in` Returns True if a sequence with the specified value is present in the object
- `not in` Returns True if a sequence with the specified value is not present in the object

````python
x = 'Hello world'
y = {1:'a', 2:'b'}

print('H' in x)
print('hello' not in x)
print(1 in y)
print('a' in y) # False because 'in' checks keys, not values in dictionaries
````

### 7. Bitwise Operators
Bitwise operators are used to perform operations on integers at the binary (bit) level. Instead of treating a number as a single value, it treats the number as a sequence of bits (0s and 1s).

These operations are very fast because they are directly supported by the processor. In Python, you can view the binary representation of a number using the built-in `bin()` function.

````python
a = 10  # Binary: 1010
b = 4   # Binary: 0100

print('a in binary:', bin(a))
print('b in binary:', bin(b))
````

#### 1. Bitwise AND (`&`)
Compares each bit of the first operand to the corresponding bit of the second operand. If **both bits are 1**, the corresponding result bit is set to 1. Otherwise, it is set to 0.

```text
  1010 (10)
& 0100 (4)
  ----
  0000 (0)
```

````python
print('a & b :', a & b) # Output: 0
````

#### 2. Bitwise OR (`|`)
Compares each bit. If **at least one of the bits is 1**, the result bit is 1. If both are 0, it is 0.

```text
  1010 (10)
| 0100 (4)
  ----
  1110 (14)
```

````python
# print('a | b :', a | b) # Output: 14
print(15 | 4)
````

#### 3. Bitwise XOR (`^`)
Compares each bit. If the bits are **different**, the result bit is 1. If they are the same (both 0 or both 1), the result is 0.

```text
  1010 (10)
^ 0100 (4)
  ----
  1110 (14)
```

````python
print('a ^ b :', a ^ b) # Output: 14
````

#### 4. Bitwise NOT (`~`)
This is a unary operator (takes only one operand). It flips every bit (0 becomes 1, and 1 becomes 0).

**Deep Dive:** In Python, integers have arbitrary precision, and the computer uses *Two's Complement* to represent negative numbers. The formula for bitwise NOT is `~x = -(x + 1)`.

```text
  1010 (10)
~ 
  ----
 -(1010 + 1) = -11 = -1011 in binary math
```

````python
# print('~a :', ~a) # Output: -11
# print('~b :', ~b) # Output: -5
print(~4)
````

#### 5. Bitwise Left Shift (`<<`)
Shifts the bits of the number to the left by the specified number of positions. Zeroes are added on the right. 

**Deep Dive:** Mathematically, shifting a number to the left by `n` positions is equivalent to multiplying the number by $2^n$.

```text
  1010 (10) << 2
  ----
  101000 (40)
```

````python
print('a << 2 :', a << 2) # Output: 40  (10 * 2^2)
````

#### 6. Bitwise Right Shift (`>>`)
Shifts the bits of the number to the right by the specified number of positions. The rightmost bits are discarded.

**Deep Dive:** Mathematically, shifting a number to the right by `n` positions is equivalent to integer (floor) division by $2^n$.

```text
  1010 (10) >> 2
  ----
  0010 (2)
```

````python
print('a >> 2 :', a >> 2) # Output: 2  (10 // 2^2)
# print(4>>2)
````

#### Bit Manipulation (Bit masking)

````python
# create all subsets of a given set s={a,b,c} 

s = ['a', 'b', 'c']
n = len(s)

for mask in range(1 << n):   # 0 to 7
    subset = []

    for i in range(n):
        # check if i-th bit is set
        if mask & (1 << i):
            subset.append(s[i])

    print(subset)
````

````python
# Count differing bits between two numbers.
# Example:
# 1 = 001
# 4 = 100
# Different bits = 2

a=1
b=4

x= a^b
c=0
while x :
    c+=x&1
    x>>=1
print(c)
````

````python
# power of 2

# A power of 2 has only one set bit.
# Example:
# 8 = 1000 
n=8
if n <=1:
    print("false")
if n&(n-1)==0:
    print('power of 2')
````
