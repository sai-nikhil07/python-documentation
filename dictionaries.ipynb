## What is a Dictionary?

A dictionary is a collection of **key-value pairs**. Think of it like a real-world dictionary — the **word** is the key and the **definition** is the value. You look up a word (key) to find its meaning (value).

### Real-World Analogy

Imagine a contact book on your phone:

| Name (Key) | Phone Number (Value) |
|------------|---------------------|
| Rahul | 9876543210 |
| Priya | 9123456789 |
| Amit | 9988776655 |

You search by the **name** (key) to get the **phone number** (value). That is exactly how a Python dictionary works.

### Where are Dictionaries Used in Real Life?

- **User profiles** — storing username, email, age as key-value pairs
- **Configuration files** — app settings like `{"theme": "dark", "language": "en"}`
- **JSON data** — almost all web APIs return data as dictionaries
- **Counting** — counting word frequency in a text
- **Caching** — storing already computed results to avoid recalculation

## 1. Dictionary Properties

Before we start creating dictionaries, let us understand what makes them special.

| Property | What it means |
|----------|---------------|
| **Ordered** | From Python 3.7+, dictionaries maintain the order in which items were inserted. Earlier versions did not guarantee this. |
| **Mutable** | You can add, change, or remove key-value pairs after creation. The dictionary can be modified in place. |
| **Unique Keys** | Each key in a dictionary must be unique. If you use the same key twice, the second value overwrites the first. |
| **Key-Value Structure** | Data is stored as pairs — every key maps to exactly one value. |
| **Fast Lookup** | Dictionaries use hashing internally, making key lookups extremely fast — O(1) on average. |

````python
# Property 1: Ordered — items stay in insertion order
student = {"name": "Rahul", "age": 21, "city": "Hyderabad"}
print(student)  # always prints in the same order as inserted
````

````python
# Property 2: Mutable — we can change it after creation
student = {"name": "Rahul", "age": 21}
student["age"] = 22  # modifying existing value
student["city"] = "Hyderabad"  # adding new key-value pair
print(student)
````

````python
# Property 3: Unique Keys — duplicate keys are NOT allowed
# If you repeat a key, the last value wins
scores = {"math": 90, "science": 85, "math": 95}
print(scores)  # math will be 95, not 90
````

````python
# Property 4: Fast Lookup
# Checking if a key exists is almost instant, even in huge dictionaries
phonebook = {"Rahul": "9876543210", "Priya": "9123456789"}
print("Rahul" in phonebook)  # True — found instantly via hashing
````

## 2. Creating Dictionaries

There are multiple ways to create a dictionary in Python.

| Method | Syntax | When to use |
|--------|--------|-------------|
| Curly braces `{}` | `{"key": value}` | Most common and readable way |
| `dict()` constructor | `dict(key=value)` | When keys are simple strings |
| `dict()` with list of tuples | `dict([("a", 1), ("b", 2)])` | When building from paired data |
| Empty dictionary | `{}` or `dict()` | When you want to add items later |
| `dict.fromkeys()` | `dict.fromkeys(["a", "b"], 0)` | When all keys share the same initial value |
| Dictionary comprehension | `{x: x**2 for x in range(5)}` | When generating from a pattern or loop |

````python
# Method 1: Using curly braces { } — most common way
student = {"name": "Rahul", "age": 21, "grade": "A"}
print(student)
print(type(student))
````

````python
# Method 2: Using dict() constructor
# Note: keys are written WITHOUT quotes and WITHOUT spaces around =
student = dict(name="Rahul", age=21, grade="A")
print(student)
````

````python
# Method 3: Using dict() with a list of tuples
# Useful when you have paired data (e.g., from a zip or database query)
pairs = [("name", "Priya"), ("age", 22), ("city", "Mumbai")]
student = dict(pairs)
print(student)
````

````python
# Method 4: Creating from two separate lists using zip()
keys = ["name", "age", "city"]
values = ["Amit", 20, "Delhi"]
student = dict(zip(keys, values))
print(student)
````

````python
# Method 5: Empty dictionary — add items later
empty1 = {}
empty2 = dict()
print(f"empty1: {empty1}, type: {type(empty1)}")
print(f"empty2: {empty2}, type: {type(empty2)}")

# CAUTION: {} creates an empty dict, NOT an empty set!
# To create an empty set, use set()
````

## 3. Accessing Values

You can retrieve values from a dictionary using the key. There are two main approaches:

| Method | Syntax | If key is missing |
|--------|--------|------------------|
| Square brackets `[]` | `dict[key]` | Raises `KeyError` ❌ |
| `get()` method | `dict.get(key)` | Returns `None` (or a default you specify) ✅ |

**Rule of thumb:** Use `get()` when you are unsure if the key exists. Use `[]` when you are certain the key exists and want an error if it does not.

````python
student = {"name": "Rahul", "age": 21, "city": "Hyderabad"}

# Method 1: Using square brackets []
print(student["name"])   # Rahul
print(student["age"])    # 21
````

````python
# What happens if the key does not exist? — KeyError!
try:
    print(student["phone"])
except KeyError as e:
    print(f"KeyError: {e} — this key does not exist in the dictionary")
````

````python
# Method 2: Using get() — safer approach
print(student.get("name"))    # Rahul
print(student.get("phone"))   # None — no error, just returns None
````

````python
# get() with a default value
# If the key is missing, return the default instead of None
phone = student.get("phone", "Not Available")
print(phone)  # Not Available

# This is super useful in real apps — e.g., getting user preferences
theme = student.get("theme", "light")  # default theme
print(f"Theme: {theme}")
````

## 4. Adding Data

To add a new key-value pair, simply assign a value to a new key using square bracket notation.

```python
dict[new_key] = value
```

If the key already exists, this will **overwrite** the value (which is updating, covered in the next section).

````python
student = {"name": "Rahul", "age": 21}
print("Before:", student)

# Adding new key-value pairs
student["city"] = "Hyderabad"
student["grade"] = "A"
student["phone"] = "9876543210"

print("After:", student)
````

````python
# Adding different types of values
student["marks"] = [85, 90, 78]          # list as value
student["address"] = {"street": "MG Road", "pin": "500001"}  # dict as value
student["is_active"] = True              # boolean as value

print(student)
````

## 5. Updating Values

There are two main ways to update values in a dictionary:

| Method | Use case |
|--------|----------|
| `dict[key] = new_value` | Update a single key's value |
| `dict.update()` | Update multiple keys at once, or merge another dict into this one |

````python
student = {"name": "Rahul", "age": 21, "city": "Hyderabad"}

# Method 1: Direct assignment — update single value
student["age"] = 22
print("After updating age:", student)
````

````python
# Method 2: update() — update multiple values at once
student.update({"age": 23, "city": "Bangalore", "grade": "A+"})
print("After update():", student)

# Notice: age and city got updated, grade got added (it was new)
````

````python
# update() can also accept keyword arguments
student.update(phone="9876543210", email="rahul@email.com")
print("After keyword update:", student)
````

````python
# update() with a list of tuples
student.update([("semester", 4), ("branch", "CSE")])
print("After tuple update:", student)
````

## 6. Removing Elements

Python provides several ways to remove items from a dictionary:

| Method | What it does | Returns |
|--------|-------------|--------|
| `pop(key)` | Removes the item with the specified key | The removed value |
| `pop(key, default)` | Same as pop, but returns default if key not found | The removed value or default |
| `popitem()` | Removes the **last inserted** key-value pair | A tuple `(key, value)` |
| `del dict[key]` | Removes the item with the specified key | Nothing (raises KeyError if missing) |
| `del dict` | Deletes the entire dictionary variable | Nothing |
| `clear()` | Removes all items, leaving an empty dictionary | Nothing |

````python
student = {"name": "Rahul", "age": 21, "city": "Hyderabad", "grade": "A"}

# pop() — removes and returns the value
removed_value = student.pop("grade")
print(f"Removed: {removed_value}")
print(f"After pop: {student}")
````

````python
# pop() with default — safe removal (no error if key missing)
removed = student.pop("phone", "Key not found")
print(f"Removed: {removed}")  # Key not found — no crash!
````

````python
# popitem() — removes the LAST inserted item
student = {"name": "Rahul", "age": 21, "city": "Hyderabad"}
last_item = student.popitem()
print(f"Removed last item: {last_item}")  # ('city', 'Hyderabad')
print(f"After popitem: {student}")
````

````python
# del — removes a specific key
student = {"name": "Rahul", "age": 21, "city": "Hyderabad"}
del student["age"]
print(f"After del: {student}")

# del with a key that does not exist — raises KeyError
try:
    del student["phone"]
except KeyError as e:
    print(f"KeyError: {e}")
````

````python
# clear() — empties the dictionary but keeps the variable
student = {"name": "Rahul", "age": 21, "city": "Hyderabad"}
student.clear()
print(f"After clear: {student}")  # {}
print(f"Type: {type(student)}")   # still a dict
````

````python
# del dict — deletes the entire variable
temp = {"x": 1, "y": 2}
del temp

try:
    print(temp)  # will crash — variable no longer exists
except NameError as e:
    print(f"NameError: {e}")
````

### Quick Comparison: When to Use Which?

| Scenario | Best method |
|----------|-------------|
| Remove a key and use its value | `pop()` |
| Remove a key safely (might not exist) | `pop(key, default)` |
| Remove the most recently added item | `popitem()` |
| Remove a key, don't need the value | `del` |
| Empty everything but keep the dict | `clear()` |
| Destroy the variable entirely | `del dict` |

## 7. Dictionary Methods — keys(), values(), items()

These three methods give you **view objects** of the dictionary's contents. They are dynamic — if the dictionary changes, the views reflect those changes.

| Method | Returns |
|--------|--------|
| `keys()` | All the keys as a `dict_keys` view |
| `values()` | All the values as a `dict_values` view |
| `items()` | All key-value pairs as a `dict_items` view (tuples) |

````python
student = {"name": "Rahul", "age": 21, "city": "Hyderabad", "grade": "A"}

# keys() — returns all keys
print("Keys:", student.keys())
print("Type:", type(student.keys()))
````

````python
# values() — returns all values
print("Values:", student.values())

# Convert to list if you need indexing
values_list = list(student.values())
print("First value:", values_list[0])
````

````python
# items() — returns all key-value pairs as tuples
print("Items:", student.items())

# Each item is a tuple: (key, value)
for item in student.items():
    print(item, "→ type:", type(item))
````

````python
# View objects are DYNAMIC — they update when the dict changes
student = {"name": "Rahul", "age": 21}
keys_view = student.keys()
print("Before:", keys_view)

student["city"] = "Hyderabad"
print("After adding 'city':", keys_view)  # automatically updated!
````

## 8. Looping Through a Dictionary

There are three common ways to loop through a dictionary:

| Loop type | What you get |
|-----------|-------------|
| `for key in dict` | Keys only |
| `for value in dict.values()` | Values only |
| `for key, value in dict.items()` | Both key and value |

````python
student = {"name": "Rahul", "age": 21, "city": "Hyderabad", "grade": "A"}

# Loop 1: Loop through KEYS (default behavior)
print("--- Keys ---")
for key in student:
    print(key)
````

````python
# Loop 2: Loop through VALUES
print("--- Values ---")
for value in student.values():
    print(value)
````

````python
# Loop 3: Loop through KEY-VALUE pairs — most useful!
print("--- Key-Value Pairs ---")
for key, value in student.items():
    print(f"{key}: {value}")
````

````python
# Practical example: formatting student data for display
student = {"name": "Rahul", "age": 21, "city": "Hyderabad", "grade": "A"}

print("=" * 30)
print("   STUDENT PROFILE")
print("=" * 30)
for key, value in student.items():
    print(f"  {key.upper():>10} : {value}")
print("=" * 30)
````

````python
# Using enumerate with items() to get index too
for index, (key, value) in enumerate(student.items(), start=1):
    print(f"{index}. {key} → {value}")
````

## 9. Checking if a Key Exists

Use the `in` keyword to check if a key exists in a dictionary. This is important to avoid `KeyError` when accessing values.

```python
key in dict        # True if key exists
key not in dict    # True if key does NOT exist
```

**Note:** `in` checks **keys only**, NOT values.

````python
student = {"name": "Rahul", "age": 21, "city": "Hyderabad"}

# Check if a key exists
print("name" in student)    # True
print("phone" in student)   # False
print("phone" not in student)  # True
````

````python
# IMPORTANT: 'in' checks KEYS, not values!
print("Rahul" in student)       # False — "Rahul" is a value, not a key
print("Rahul" in student.values())  # True — now we are checking values
````

````python
# Practical use: safe access pattern
student = {"name": "Rahul", "age": 21}

key = "phone"
if key in student:
    print(f"{key}: {student[key]}")
else:
    print(f"'{key}' not found in the dictionary")
````

## 10. Nested Dictionaries

A nested dictionary is a dictionary inside another dictionary. This is extremely common in real-world programming — think of JSON data from APIs, configuration files, or database records.

### Real-World Example
A school database where each student has their own dictionary of details.

````python
# Nested dictionary — students database
students = {
    "student1": {
        "name": "Rahul",
        "age": 21,
        "marks": {"math": 90, "science": 85, "english": 78}
    },
    "student2": {
        "name": "Priya",
        "age": 22,
        "marks": {"math": 95, "science": 88, "english": 92}
    },
    "student3": {
        "name": "Amit",
        "age": 20,
        "marks": {"math": 70, "science": 75, "english": 80}
    }
}

print(students)
````

````python
# Accessing nested values — chain the keys
print(students["student1"]["name"])              # Rahul
print(students["student2"]["marks"]["math"])      # 95
print(students["student3"]["age"])                # 20
````

````python
# Modifying nested values
students["student1"]["marks"]["math"] = 95
print("Updated math score:", students["student1"]["marks"]["math"])
````

````python
# Adding a new student
students["student4"] = {
    "name": "Sneha",
    "age": 21,
    "marks": {"math": 88, "science": 92, "english": 85}
}
print(f"Total students: {len(students)}")
````

````python
# Looping through nested dictionary
for student_id, details in students.items():
    print(f"\n{student_id}:")
    for key, value in details.items():
        if isinstance(value, dict):
            print(f"  {key}:")
            for subject, mark in value.items():
                print(f"    {subject}: {mark}")
        else:
            print(f"  {key}: {value}")
````

````python
# Practical: Find the topper in math
math_topper = max(students.items(), key=lambda x: x[1]["marks"]["math"])
print(f"Math topper: {math_topper[1]['name']} with {math_topper[1]['marks']['math']} marks")
````

## 11. Dictionary Comprehension

Dictionary comprehension is a concise way to create dictionaries using a single line of code. It follows the same idea as list comprehension but produces key-value pairs.

### Syntax

```python
{key_expression: value_expression for item in iterable}
{key_expression: value_expression for item in iterable if condition}
```

````python
# Basic: create a dictionary of squares
squares = {x: x ** 2 for x in range(1, 6)}
print(squares)  # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
````

````python
# With condition: only even number squares
even_squares = {x: x ** 2 for x in range(1, 11) if x % 2 == 0}
print(even_squares)  # {2: 4, 4: 16, 6: 36, 8: 64, 10: 100}
````

````python
# From two lists — names and scores
names = ["Rahul", "Priya", "Amit", "Sneha"]
scores = [85, 92, 78, 88]

result = {name: score for name, score in zip(names, scores)}
print(result)
````

````python
# Filter: only students who scored above 80
toppers = {name: score for name, score in zip(names, scores) if score > 80}
print("Toppers:", toppers)
````

````python
# Transform: convert string values to uppercase
student = {"name": "rahul", "city": "hyderabad", "grade": "a"}
upper_student = {k: v.upper() if isinstance(v, str) else v for k, v in student.items()}
print(upper_student)
````

````python
# Swap keys and values
original = {"a": 1, "b": 2, "c": 3}
swapped = {v: k for k, v in original.items()}
print(f"Original: {original}")
print(f"Swapped:  {swapped}")
````

````python
# Nested comprehension: multiplication table
table = {i: {j: i * j for j in range(1, 6)} for i in range(2, 5)}
for num, products in table.items():
    print(f"Table of {num}: {products}")
````

## 12. Sorting a Dictionary

Dictionaries themselves cannot be sorted in-place (they maintain insertion order). But you can create a **new dictionary** with items in sorted order.

| Goal | Code |
|------|------|
| Sort by key | `dict(sorted(d.items()))` |
| Sort by value | `dict(sorted(d.items(), key=lambda x: x[1]))` |
| Sort in reverse | Add `reverse=True` |

````python
marks = {"Priya": 92, "Amit": 78, "Rahul": 85, "Sneha": 88}

# Sort by key (alphabetical order of names)
sorted_by_key = dict(sorted(marks.items()))
print("Sorted by key:", sorted_by_key)
````

````python
# Sort by value (ascending marks)
sorted_by_value = dict(sorted(marks.items(), key=lambda x: x[1]))
print("Sorted by value (asc):", sorted_by_value)
````

````python
# Sort by value (descending — highest marks first)
sorted_desc = dict(sorted(marks.items(), key=lambda x: x[1], reverse=True))
print("Sorted by value (desc):", sorted_desc)
````

````python
# Sort by key in reverse order
sorted_key_desc = dict(sorted(marks.items(), reverse=True))
print("Sorted by key (desc):", sorted_key_desc)
````

````python
# Using lambda for custom sorting — sort by the length of the key name
sorted_by_name_length = dict(sorted(marks.items(), key=lambda x: len(x[0])))
print("Sorted by name length:", sorted_by_name_length)
````

````python
# Get only sorted keys or sorted values
print("Sorted keys:", sorted(marks.keys()))
print("Sorted values:", sorted(marks.values()))
````

## 13. Copying a Dictionary

This is a very important concept that trips up many beginners.

| Method | What happens | Type of copy |
|--------|-------------|-------------|
| `=` (assignment) | Both variables point to the **same** dictionary | **No copy** (alias) |
| `.copy()` | Creates a new dictionary with the same items | **Shallow copy** |
| `copy.deepcopy()` | Creates a completely independent copy (even nested objects) | **Deep copy** |

### Why does this matter?

With `=`, changing one variable changes the other too — because they share the same memory!

````python
# Problem: Assignment (=) does NOT create a copy!
original = {"name": "Rahul", "age": 21}
alias = original  # both point to the SAME dictionary

alias["age"] = 99
print("original:", original)  # age is 99! Original got changed too!
print("alias:", alias)
print("Same object?", original is alias)  # True
````

````python
# Solution: Use .copy() for a shallow copy
original = {"name": "Rahul", "age": 21}
copied = original.copy()

copied["age"] = 99
print("original:", original)  # age is still 21 ✅
print("copied:", copied)      # age is 99
print("Same object?", original is copied)  # False
````

````python
# Also works: dict() constructor creates a copy
original = {"name": "Rahul", "age": 21}
copied = dict(original)
copied["age"] = 99
print("original:", original)  # unchanged ✅
````

````python
# WARNING: Shallow copy has a limitation with nested objects!
original = {"name": "Rahul", "marks": [85, 90, 78]}
shallow = original.copy()

shallow["marks"].append(95)  # modifying the nested list
print("original:", original)  # marks changed! [85, 90, 78, 95] 😱
print("shallow:", shallow)
# The inner list is still shared between both dictionaries!
````

````python
# Solution: Use deepcopy for nested structures
import copy

original = {"name": "Rahul", "marks": [85, 90, 78]}
deep = copy.deepcopy(original)

deep["marks"].append(95)
print("original:", original)  # unchanged ✅ [85, 90, 78]
print("deep:", deep)          # [85, 90, 78, 95]
````

## 14. Merging Dictionaries

Combining two or more dictionaries is a common task. Python offers several ways:

| Method | Python Version | Syntax |
|--------|---------------|--------|
| `\|` operator | 3.9+ | `d1 \| d2` |
| `\|=` operator | 3.9+ | `d1 \|= d2` (in-place) |
| `update()` | All | `d1.update(d2)` |
| `{**d1, **d2}` | 3.5+ | Unpacking with `**` |

**Note:** If both dictionaries have the same key, the value from the **second** dictionary wins.

````python
dict1 = {"name": "Rahul", "age": 21}
dict2 = {"city": "Hyderabad", "grade": "A"}

# Method 1: Using | operator (Python 3.9+)
merged = dict1 | dict2
print("Merged with |:", merged)
````

````python
# Method 2: Using |= for in-place merge (Python 3.9+)
dict1 = {"name": "Rahul", "age": 21}
dict2 = {"city": "Hyderabad", "grade": "A"}

dict1 |= dict2  # dict1 is modified in place
print("After |=", dict1)
````

````python
# Method 3: Using update() — modifies the original dict
dict1 = {"name": "Rahul", "age": 21}
dict2 = {"city": "Hyderabad", "grade": "A"}

dict1.update(dict2)
print("After update():", dict1)
````

````python
# Method 4: Using ** unpacking — creates a new dict
dict1 = {"name": "Rahul", "age": 21}
dict2 = {"city": "Hyderabad", "grade": "A"}

merged = {**dict1, **dict2}
print("Merged with **:", merged)
````

````python
# When keys overlap — last value wins
defaults = {"theme": "light", "language": "en", "font_size": 14}
user_prefs = {"theme": "dark", "font_size": 18}

final_settings = defaults | user_prefs
print("Final settings:", final_settings)
# theme = "dark" (user overrides default)
# font_size = 18 (user overrides default)
# language = "en" (no user override, default stays)
````

````python
# Merging more than 2 dictionaries
a = {"x": 1}
b = {"y": 2}
c = {"z": 3}

merged = a | b | c
print("Three-way merge:", merged)

# Or with unpacking
merged = {**a, **b, **c}
print("Three-way ** merge:", merged)
````

## 15. setdefault() Method

`setdefault()` is a handy method that does **two things** in one call:

1. If the key **exists** → returns its value (does NOT change anything)
2. If the key **does not exist** → inserts the key with a default value and returns it

```python
dict.setdefault(key, default_value)
```

This is very useful when building dictionaries on the fly, especially when grouping data.

````python
student = {"name": "Rahul", "age": 21}

# Key exists — returns existing value, does NOT change it
name = student.setdefault("name", "Unknown")
print(f"name: {name}")       # Rahul (not "Unknown")
print(f"dict: {student}")    # unchanged
````

````python
# Key does NOT exist — inserts it with the default value
city = student.setdefault("city", "Hyderabad")
print(f"city: {city}")       # Hyderabad
print(f"dict: {student}")    # city has been added!
````

````python
# Practical use: Grouping items into lists
# Without setdefault (verbose way)
words = ["apple", "ant", "ball", "bat", "cat", "car", "ant"]
grouped = {}

for word in words:
    first_letter = word[0]
    if first_letter not in grouped:
        grouped[first_letter] = []
    grouped[first_letter].append(word)

print("Without setdefault:", grouped)
````

````python
# With setdefault (cleaner way — same result!)
words = ["apple", "ant", "ball", "bat", "cat", "car", "ant"]
grouped = {}

for word in words:
    grouped.setdefault(word[0], []).append(word)

print("With setdefault:", grouped)
````

````python
# setdefault vs get()
# get() does NOT modify the dictionary — it only reads
# setdefault() modifies the dictionary if the key is missing

d = {"a": 1}
d.get("b", 0)          # returns 0, but does NOT add "b" to d
print("After get:", d)  # {"a": 1} — unchanged

d.setdefault("b", 0)    # returns 0, AND adds "b": 0 to d
print("After setdefault:", d)  # {"a": 1, "b": 0} — changed!
````

## 16. fromkeys() Method

`dict.fromkeys()` creates a new dictionary from a sequence of keys, all initialized to the same value.

```python
dict.fromkeys(keys, value)
```

| Parameter | Description |
|-----------|------------|
| `keys` | An iterable (list, tuple, string, etc.) of keys |
| `value` | The default value for all keys (optional, defaults to `None`) |

````python
# Create a dictionary with all keys set to 0
subjects = ["math", "science", "english", "hindi"]
marks = dict.fromkeys(subjects, 0)
print(marks)  # {'math': 0, 'science': 0, 'english': 0, 'hindi': 0}
````

````python
# Without a default value — all keys get None
fields = ["name", "email", "phone"]
empty_form = dict.fromkeys(fields)
print(empty_form)  # {'name': None, 'email': None, 'phone': None}
````

````python
# Using a string as keys — each character becomes a key
char_dict = dict.fromkeys("hello", 0)
print(char_dict)  # {'h': 0, 'e': 0, 'l': 0, 'o': 0}
# Note: duplicate 'l' appears only once (keys are unique)
````

````python
# ⚠️ CAUTION: Mutable default values are shared!
# This is a common bug
keys = ["a", "b", "c"]
d = dict.fromkeys(keys, [])  # all keys share the SAME list object

d["a"].append(1)
print(d)  # {'a': [1], 'b': [1], 'c': [1]} — all changed! 😱

# Solution: Use a dictionary comprehension instead
d = {key: [] for key in keys}  # each key gets its OWN list
d["a"].append(1)
print(d)  # {'a': [1], 'b': [], 'c': []} — correct ✅
````

## 17. Valid and Invalid Keys

Not everything can be used as a dictionary key. The rule is simple:

**Dictionary keys must be hashable (and therefore immutable).**

### What does "hashable" mean?

An object is hashable if it has a hash value that never changes during its lifetime. Python uses this hash to quickly locate the key in memory.

| Type | Hashable? | Can be a key? | Example |
|------|-----------|--------------|--------|
| `int` | ✅ Yes | ✅ Yes | `{1: "one"}` |
| `float` | ✅ Yes | ✅ Yes | `{3.14: "pi"}` |
| `str` | ✅ Yes | ✅ Yes | `{"name": "Rahul"}` |
| `bool` | ✅ Yes | ✅ Yes | `{True: "yes"}` |
| `tuple` | ✅ Yes* | ✅ Yes* | `{(1, 2): "point"}` |
| `None` | ✅ Yes | ✅ Yes | `{None: "nothing"}` |
| `frozenset` | ✅ Yes | ✅ Yes | `{frozenset({1,2}): "set"}` |
| `list` | ❌ No | ❌ No | `{[1, 2]: "fail"}` → TypeError |
| `dict` | ❌ No | ❌ No | `{{}: "fail"}` → TypeError |
| `set` | ❌ No | ❌ No | `{{1, 2}: "fail"}` → TypeError |

*Tuples are hashable only if ALL their elements are also hashable.

````python
# Valid keys — all immutable/hashable types
valid = {
    42: "integer key",
    3.14: "float key",
    "name": "string key",
    True: "boolean key",
    (1, 2): "tuple key",
    None: "None key",
    frozenset({1, 2}): "frozenset key"
}

for key, value in valid.items():
    print(f"  {str(key):>20} → {value}")
````

````python
# Invalid keys — mutable types cause TypeError

# List as key
try:
    d = {[1, 2, 3]: "list key"}
except TypeError as e:
    print(f"List as key: {e}")

# Dict as key
try:
    d = {{"a": 1}: "dict key"}
except TypeError as e:
    print(f"Dict as key: {e}")

# Set as key
try:
    d = {{1, 2}: "set key"}
except TypeError as e:
    print(f"Set as key: {e}")
````

````python
# Tuple with mutable element inside — also INVALID!
try:
    d = {(1, [2, 3]): "tuple with list"}
except TypeError as e:
    print(f"Tuple with list inside: {e}")

# Tuple with only immutable elements — valid
d = {(1, 2, 3): "pure tuple"}
print(f"Valid: {d}")
````

````python
# Quick way to check if something is hashable
test_values = [42, "hello", [1, 2], {"a": 1}, (1, 2), {1, 2}, None, frozenset({1, 2})]

for val in test_values:
    try:
        hash(val)
        print(f"  {str(val):>20} → hashable ✅ (can be a key)")
    except TypeError:
        print(f"  {str(val):>20} → NOT hashable ❌ (cannot be a key)")
````

````python
# Gotcha: True and 1 are the same key! (also False and 0)
# Because bool is a subclass of int in Python
d = {True: "boolean", 1: "integer"}
print(d)          # {True: 'integer'} — 1 overwrote True's value
print(len(d))     # 1 — they are treated as the same key!

print(hash(True) == hash(1))  # True
print(True == 1)              # True
````

## 18. Frequency Counter Using Dictionary

One of the most common dictionary use cases is counting how many times each item appears in a collection. This is called a **frequency counter**.

We will look at three approaches:

1. **Manual counting** with a plain dictionary
2. Using `collections.Counter` (built-in, best way)
3. Using `dict.get()` for cleaner manual counting

### Approach 1: Manual Counting with a Plain Dictionary

````python
# Count character frequency in a string
text = "hello world"
freq = {}

for char in text:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1

print("Character frequency:", freq)
````

````python
# Cleaner manual counting using get()
text = "hello world"
freq = {}

for char in text:
    freq[char] = freq.get(char, 0) + 1

print("With get():", freq)
````

````python
# Count word frequency in a sentence
sentence = "the cat sat on the mat the cat is fat"
words = sentence.split()
word_count = {}

for word in words:
    word_count[word] = word_count.get(word, 0) + 1

print("Word frequency:")
for word, count in sorted(word_count.items(), key=lambda x: x[1], reverse=True):
    print(f"  '{word}' → {count} times")
````

### Approach 2: Using `collections.Counter` (Recommended)

`Counter` is a special dictionary subclass designed specifically for counting. It is the **fastest and most Pythonic** way to count things.

| Feature | Description |
|---------|------------|
| Auto-counting | Pass any iterable and it counts automatically |
| `most_common(n)` | Returns the n most frequent items |
| Arithmetic | Supports `+`, `-`, `&`, `\|` between Counters |
| Missing keys | Returns `0` for missing keys (no KeyError!) |

````python
from collections import Counter

# Count characters
text = "hello world"
char_count = Counter(text)
print("Character count:", char_count)
````

````python
# Count words
sentence = "the cat sat on the mat the cat is fat"
word_count = Counter(sentence.split())
print("Word count:", word_count)
````

````python
# most_common(n) — get the top n most frequent items
print("Top 3 words:", word_count.most_common(3))
print("Least common:", word_count.most_common()[-1])  # last item
````

````python
# Counter returns 0 for missing keys — never raises KeyError!
print("Count of 'dog':", word_count["dog"])  # 0, not KeyError
````

````python
# Counter arithmetic — combining and subtracting counts
sales_jan = Counter({"apples": 10, "bananas": 5, "oranges": 8})
sales_feb = Counter({"apples": 7, "bananas": 12, "grapes": 3})

# Total sales
total = sales_jan + sales_feb
print("Total sales:", total)

# Difference
diff = sales_jan - sales_feb
print("Jan - Feb:", diff)  # only shows positive counts
````

````python
# Count items in a list
votes = ["Alice", "Bob", "Alice", "Charlie", "Bob", "Alice", "Bob", "Bob"]
result = Counter(votes)

print("Election results:")
for candidate, count in result.most_common():
    bar = "█" * count
    print(f"  {candidate:>10}: {bar} ({count} votes)")

winner = result.most_common(1)[0]
print(f"\nWinner: {winner[0]} with {winner[1]} votes! 🎉")
````

### Approach 3: Using `defaultdict` from collections

Another useful tool from `collections`. It automatically creates a default value when you access a missing key — no need for `if` checks or `get()`.

````python
from collections import defaultdict

# defaultdict(int) — default value is 0
text = "hello world"
freq = defaultdict(int)

for char in text:
    freq[char] += 1  # no need to check if key exists!

print(dict(freq))
````

````python
# defaultdict(list) — great for grouping!
students = [
    ("CSE", "Rahul"), ("ECE", "Priya"), ("CSE", "Amit"),
    ("ECE", "Sneha"), ("CSE", "Nikhil"), ("ME", "Ravi")
]

by_branch = defaultdict(list)
for branch, name in students:
    by_branch[branch].append(name)

print("Students by branch:")
for branch, names in by_branch.items():
    print(f"  {branch}: {names}")
````

## 19. Useful Built-in Functions with Dictionaries

Here are some built-in Python functions that work well with dictionaries:

| Function | What it does |
|----------|-------------|
| `len(d)` | Number of key-value pairs |
| `max(d)` / `min(d)` | Largest/smallest key |
| `sum(d.values())` | Sum of all values (if numeric) |
| `any(d.values())` | True if any value is truthy |
| `all(d.values())` | True if all values are truthy |

````python
marks = {"math": 90, "science": 85, "english": 78, "hindi": 92}

print(f"Total subjects: {len(marks)}")
print(f"Total marks: {sum(marks.values())}")
print(f"Average: {sum(marks.values()) / len(marks):.1f}")
print(f"Highest scoring subject: {max(marks, key=marks.get)}")
print(f"Lowest scoring subject: {min(marks, key=marks.get)}")
print(f"Max marks: {max(marks.values())}")
print(f"Min marks: {min(marks.values())}")
````

````python
# any() and all()
status = {"task1": True, "task2": False, "task3": True}

print(f"Any task done? {any(status.values())}")   # True
print(f"All tasks done? {all(status.values())}")   # False
````

## 20. Common Mistakes to Avoid

Here are the most common mistakes beginners make with dictionaries:

````python
# Mistake 1: Modifying a dictionary while looping over it
d = {"a": 1, "b": 2, "c": 3}

# ❌ This will cause RuntimeError
try:
    for key in d:
        if d[key] < 3:
            del d[key]
except RuntimeError as e:
    print(f"Error: {e}")

# ✅ Correct: iterate over a copy of keys
d = {"a": 1, "b": 2, "c": 3}
for key in list(d.keys()):
    if d[key] < 3:
        del d[key]
print("After safe deletion:", d)
````

````python
# Mistake 2: Using = instead of .copy()
original = {"a": 1}
wrong_copy = original          # NOT a copy!
correct_copy = original.copy() # actual copy

wrong_copy["b"] = 2
print("original after wrong copy:", original)  # modified! 😱

correct_copy["c"] = 3
print("original after correct copy:", original)  # safe ✅
````

````python
# Mistake 3: Assuming dict.fromkeys() with mutable default is safe
# Already covered in section 16 — each key shares the same object!
# Always use comprehension: {k: [] for k in keys}
print("See section 16 for the fromkeys() mutable default gotcha!")
````

````python
# Mistake 4: Confusing {} with set()
empty_dict = {}       # this is a dict, NOT a set
empty_set = set()     # this is a set

print(type(empty_dict))  # <class 'dict'>
print(type(empty_set))   # <class 'set'>
````
