## 1. What is a Lambda Function?

In everyday programming, you often need a small function that does one simple thing — maybe double a number, or check if something is positive. Writing a full `def` block for such tiny tasks can feel like overkill.

Python gives you a shortcut for this: the **lambda function**.

A lambda function is a **small, anonymous, one-line function**. "Anonymous" means it does not have a name (unless you assign it to a variable). It is designed for cases where you need a quick, throwaway piece of logic without the formality of defining a full function.

### Real-World Analogy

Think of a regular function like a named recipe in a cookbook — it has a title, ingredients, and steps. You write it down because you will use it many times.

A lambda is like a quick mental note you scribble on a sticky note: "add 10 to the price." You do not give it a formal name because you only need it once, right now.

### Where are Lambdas Used in Real Life?

- **E-commerce systems** — applying a discount to every price in a product list
- **Data analysis** — transforming a column in a spreadsheet (e.g., converting currency)
- **Sorting** — ordering a list of employees by their joining date or salary
- **Filtering** — keeping only the active users from a database result
- **Pandas / data science** — applying custom logic row-by-row in a DataFrame

---
## 2. Syntax Breakdown

The structure of a lambda function is:

```
lambda arguments : expression
```

Let us break each part down:

| Part | What it means |
|------|---------------|
| `lambda` | The keyword that tells Python: "this is an anonymous function" |
| `arguments` | The inputs the function receives (just like parameters in `def`) |
| `:` | Separates the arguments from the logic |
| `expression` | A single line of logic whose result is automatically returned |

### Important Rules

- A lambda can only contain **one expression** — it cannot have multiple lines, loops, or if-else blocks as statements
- It **automatically returns** the result of the expression — there is no `return` keyword
- It can take **zero, one, or many arguments**
- You can assign a lambda to a variable to give it a name and reuse it

````python
# The simplest lambda — takes one input, returns it doubled
double = lambda x: x * 2

print(double(5))   # 10
print(double(21))  # 42

# Lambda with two inputs
add = lambda x, y: x + y
print(add(3, 7))   # 10

# Lambda with no input
greet = lambda: "Hello!"
print(greet())     # Hello!

# Lambda with a default argument value
power = lambda x, n=2: x ** n
print(power(4))    # 16  (uses n=2 by default)
print(power(2, 8)) # 256 (overrides n with 8)
````

## 3. Lambda vs Regular Functions

A lambda is not a replacement for `def`. It is a companion — each has its place.

Here is a side-by-side comparison of the same logic written both ways:

````python
# Regular function
def apply_gst(price):
    return round(price * 1.18, 2)

# Equivalent lambda
apply_gst_lambda = lambda price: round(price * 1.18, 2)

print(apply_gst(1000))        # 1180.0
print(apply_gst_lambda(1000)) # 1180.0
````

Both produce the exact same result. So when should you use which?

### Use `def` when:

- The function will be **reused** in many places across your code
- The logic is **complex** (multiple steps, conditions, loops)
- You want to write a **docstring** to explain what it does
- You need to **raise exceptions** or handle errors inside it
- You want your code to be **easy to read and test**

### Use `lambda` when:

- The function is **short and used once** (typically as an argument to another function)
- You are passing it to `map()`, `filter()`, `sorted()`, or `reduce()`
- Writing a full `def` would interrupt the flow and add unnecessary lines

### The Golden Rule

> If someone reading your code has to stop and think about what the lambda does, write a `def` instead.

Lambda is a tool for **clarity and brevity** — not for cleverness.

## 4. `map()` — Transforming Every Item in a List

### What is map()?

`map()` takes a function and applies it to **every single item** in a list (or any iterable). It transforms the entire collection without writing a loop.

```
map(function, iterable)
```

The result is a **map object** (a lazy iterator). You wrap it in `list()` to get a regular list back.

### Real-World Analogy

Imagine a conveyor belt in a factory. Each item comes in, goes through the same machine, and comes out transformed. `map()` is that machine — it processes every item the same way.

### When is map() useful?

- An HR system needs to **apply a 10% salary raise** to every employee in a list
- A shopping app needs to **add 18% GST** to every product price before displaying
- A data pipeline needs to **convert all temperatures** from Celsius to Fahrenheit
- A content platform needs to **capitalize all usernames** fetched from a database

````python
# Real-World Example 1: Apply GST to a list of product prices
# An e-commerce platform stores base prices. Before showing to users,
# it needs to add 18% GST to every price.

base_prices = [1000, 2500, 850, 4200, 600]

prices_with_gst = list(map(lambda price: round(price * 1.18, 2), base_prices))

print("Base prices :", base_prices)
print("With 18% GST:", prices_with_gst)
````

````python
# Real-World Example 2: Standardize customer names from a database
# A CRM system pulled customer names that are inconsistently formatted.
# Before sending emails, the system needs to clean and capitalize them.

raw_names = ["  alice sharma  ", "BOB VERMA", "charlie das", "  DIANA NAIR"]

clean_names = list(map(lambda name: name.strip().title(), raw_names))

print("Raw names  :", raw_names)
print("Clean names:", clean_names)
````

````python
# Real-World Example 3: Apply annual salary hike to all employees
# The HR department announces a 12% hike. The system needs to update
# every employee's salary in one operation.

current_salaries = [45000, 72000, 90000, 60000, 110000]

updated_salaries = list(map(lambda s: round(s * 1.12), current_salaries))

for before, after in zip(current_salaries, updated_salaries):
    print(f"Before: Rs. {before:>1,}   After (12% hike): Rs. {after:>7,}")
````

### Key Point about map()

`map()` is about **transformation** — every item goes in, every item comes out, just in a different form. The number of items in the output always equals the number of items in the input. Nothing is added or removed.

If you want to also **remove** some items based on a condition — that is where `filter()` comes in.

## 5. `filter()` — Keeping Only What You Need

### What is filter()?

`filter()` takes a function that returns `True` or `False`, and applies it to every item in an iterable. It **keeps** the items where the function returns `True` and **discards** the rest.

```
filter(function, iterable)
```

Like `map()`, the result is a lazy iterator — wrap it in `list()` to see the results.

### Real-World Analogy

Think of a security checkpoint at an airport. Every passenger goes through, but only those who pass the check are allowed through. The others are turned away. `filter()` is that checkpoint.

### When is filter() useful?

- A bank needs to identify **accounts with a balance below the minimum** to send alerts
- A recruitment platform wants to shortlist **candidates with more than 5 years of experience**
- An e-commerce app needs to show only **products that are currently in stock**
- A reporting tool needs to extract **only the failed transactions** from a log

````python
# Real-World Example 1: Filter products that are in stock
# An online store fetches all products but should only display
# the ones that are available (stock > 0).

products = [
    {"name": "Laptop",     "price": 85000, "stock": 5},
    {"name": "Mouse",      "price": 1200,  "stock": 0},
    {"name": "Keyboard",   "price": 3500,  "stock": 12},
    {"name": "Monitor",    "price": 22000, "stock": 0},
    {"name": "Headphones", "price": 4500,  "stock": 8},
]

available = list(filter(lambda p: p["stock"] > 0, products))

print("Products available for purchase:")
for p in available:
    print(f"  {p['name']:15}  Rs. {p['price']:,}  (Stock: {p['stock']})")
````

````python
# Real-World Example 2: Flag suspicious transactions
# A payment system monitors transactions and flags any
# that exceed Rs. 50,000 for manual review.

transactions = [
    {"id": "TXN001", "amount": 1200,  "user": "Alice"},
    {"id": "TXN002", "amount": 87500, "user": "Bob"},
    {"id": "TXN003", "amount": 4500,  "user": "Charlie"},
    {"id": "TXN004", "amount": 65000, "user": "Diana"},
    {"id": "TXN005", "amount": 300,   "user": "Eve"},
]

LIMIT = 50000
flagged = list(filter(lambda t: t["amount"] > LIMIT, transactions))

print(f"Transactions flagged for review (amount > Rs. {LIMIT:,}):")
for t in flagged:
    print(f"  {t['id']}  {t['user']:10}  Rs. {t['amount']:,}")
````

````python
# Real-World Example 3: Shortlist job applicants by experience
# A hiring platform filters candidates who meet the minimum
# experience requirement for a senior role (5+ years).

applicants = [
    {"name": "Ravi",    "experience": 7, "role": "Senior Dev"},
    {"name": "Sneha",   "experience": 2, "role": "Junior Dev"},
    {"name": "Arjun",   "experience": 5, "role": "Mid-level"},
    {"name": "Priya",   "experience": 1, "role": "Intern"},
    {"name": "Karthik", "experience": 9, "role": "Lead Dev"},
]

shortlisted = list(filter(lambda a: a["experience"] >= 5, applicants))

print("Shortlisted candidates (5+ years experience):")
for a in shortlisted:
    print(f"  {a['name']:10}  {a['experience']} years  ({a['role']})")
````

### Key Point about filter()

`filter()` is about **selection** — you put in N items and get back fewer (or equal) items. Nothing is transformed. The items that make it through are exactly as they were.

The function you pass must return a **boolean** (`True`/`False`). This is called a **predicate**.

A quick way to remember the difference:

- `map()` — change the shape of every item (transformation)
- `filter()` — decide whether each item survives (selection)

## 6. `reduce()` — Collapsing a List Into One Value

### What is reduce()?

`reduce()` is not a built-in function in Python 3 — you need to import it from the `functools` module. It takes a function and an iterable, and **repeatedly applies the function** to accumulate all items into a single result.

```
from functools import reduce
reduce(function, iterable, initializer)
```

The function you pass must take **two arguments**: the accumulated value so far, and the next item from the list.

### How it works — Step by Step

Imagine you want to sum `[10, 20, 30, 40]` using reduce:

```
Step 1: acc = 10,  next = 20  --> 10 + 20 = 30
Step 2: acc = 30,  next = 30  --> 30 + 30 = 60
Step 3: acc = 60,  next = 40  --> 60 + 40 = 100
Final result: 100
```

It is like a rolling ball that picks up values as it goes — each step combines what was collected so far with the next item.

### Real-World Analogy

Think of a cashier at a grocery checkout. They scan each item one by one and keep a running total. After scanning all items, they announce the final bill. `reduce()` is that running total process.

### When is reduce() useful?

- Calculating **total revenue** from a list of sales records
- Finding the **cumulative product** (e.g., factorial, compound interest)
- **Merging** or combining a list of dictionaries into one
- Building a **pipeline** where each step feeds into the next

````python
from functools import reduce

# Real-World Example 1: Calculate total bill from a shopping cart
# A checkout system needs to sum up all item prices to show the total.

cart_items = [
    {"item": "Notebook",  "price": 120},
    {"item": "Pen Set",   "price": 85},
    {"item": "Backpack",  "price": 950},
    {"item": "Water Bottle", "price": 350},
    {"item": "Earphones",  "price": 1200},
]

total_bill = reduce(lambda acc, item: acc + item["price"], cart_items, 0)

print("Cart Summary:")
for c in cart_items:
    print(f"  {c['item']:15}  Rs. {c['price']}")
print("-" * 35)
print(f"  {'Total':15}  Rs. {total_bill}")
````

````python
# Real-World Example 2: Compound interest over multiple years
# A fixed deposit grows at different rates each year.
# reduce() applies each year's growth rate to the running balance.

principal = 100000  # Rs. 1 lakh
yearly_rates = [0.07, 0.08, 0.09, 0.07, 0.10]  # 5 years of interest rates

final_amount = reduce(lambda balance, rate: balance * (1 + rate), yearly_rates, principal)

print(f"Principal    : Rs. {principal:,}")
print(f"After 5 years: Rs. {final_amount:,.2f}")
print(f"Total growth : Rs. {final_amount - principal:,.2f}")
````

````python
# Real-World Example 3: Merge multiple configuration dictionaries
# In many applications, settings come from multiple sources
# (defaults, environment, user preferences). reduce() merges them all.
# Later sources override earlier ones.

default_config   = {"theme": "light",  "language": "en",    "notifications": True}
env_config       = {"language": "hi",  "debug": False}
user_config      = {"theme": "dark",   "notifications": False}

configs = [default_config, env_config, user_config]

final_config = reduce(lambda base, override: {**base, **override}, configs)

print("Final merged configuration:")
for key, value in final_config.items():
    print(f"  {key:20} : {value}")
````

### The optional `initializer` argument in reduce()

When you pass a third argument to `reduce()`, it acts as the **starting value** for the accumulation.

This is important when:
- The list might be empty (without an initializer, `reduce()` on an empty list throws an error)
- Your accumulator starts as something other than the first element (e.g., a running total that starts at 0, or a dict that starts empty)

````python
from functools import reduce

# Without initializer: first element becomes the starting value
result_a = reduce(lambda acc, x: acc + x, [10, 20, 30])
print("Without initializer:", result_a)   # 60

# With initializer: starts from the given value
result_b = reduce(lambda acc, x: acc + x, [10, 20, 30], 100)
print("With initializer 100:", result_b)  # 160

# Safe behavior with an empty list:
safe_result = reduce(lambda acc, x: acc + x, [], 0)
print("Empty list with initializer:", safe_result)  # 0 (no error)
````

## 7. Combining map(), filter(), and reduce()

### The Power of Chaining

Each of these three functions does one job well. Their real power shows when you **chain them together** into a pipeline — like an assembly line where each station processes what the previous one passed along.

A common pattern is:

```
filter() --> map() --> reduce()
  Select     Transform    Summarize
```

This maps directly to how data processing works in the real world:

1. **Filter** — narrow down to the records you care about
2. **Map** — extract or transform the relevant values
3. **Reduce** — aggregate them into a final answer

### Real-World Application: Sales Report Pipeline

Suppose a sales manager wants to know: *"What is the total revenue from orders above Rs. 1000, after applying a 10% seasonal discount?"*

This is three operations:
1. Filter: keep only orders above Rs. 1000
2. Map: apply the 10% discount to each
3. Reduce: sum them all up

````python
from functools import reduce

order_amounts = [500, 1200, 850, 3000, 1500, 400, 2200, 950, 5000]

# Step 1: Keep only orders above Rs. 1000
qualifying = filter(lambda amount: amount > 1000, order_amounts)

# Step 2: Apply 10% discount to each qualifying order
discounted = map(lambda amount: amount * 0.90, qualifying)

# Step 3: Sum up the final revenue
total_revenue = reduce(lambda acc, amount: acc + amount, discounted, 0)

print(f"Total discounted revenue from qualifying orders: Rs. {total_revenue:,.2f}")
````

````python
# Real-World Example 2: Employee bonus payout pipeline
# HR wants the total bonus budget for high-performing employees only.
# Rule: Performance rating >= 4.0 qualifies. Bonus = 15% of salary.

employees = [
    {"name": "Priya",   "salary": 80000,  "rating": 4.5},
    {"name": "Arjun",   "salary": 95000,  "rating": 3.8},
    {"name": "Divya",   "salary": 72000,  "rating": 4.9},
    {"name": "Kiran",   "salary": 60000,  "rating": 3.1},
    {"name": "Meera",   "salary": 110000, "rating": 4.7},
]

# Step 1: Filter — keep only high performers
high_performers = filter(lambda e: e["rating"] >= 4.0, employees)

# Step 2: Map — calculate each person's bonus amount
bonus_amounts = map(lambda e: e["salary"] * 0.15, high_performers)

# Step 3: Reduce — find the total bonus budget needed
total_bonus_budget = reduce(lambda acc, b: acc + b, bonus_amounts, 0)

print(f"Total bonus payout (high performers only): Rs. {total_bonus_budget:,.2f}")
````

## 8. Sorting with Lambda — Using `sorted()` and `key=`

### What is the `key` argument?

Python's `sorted()` function can sort lists, but by default it only knows how to compare simple values like numbers and strings. When you have a **list of dictionaries** or **list of tuples**, you need to tell it *what to sort by*.

That is where the `key=` argument comes in. You pass a function that tells `sorted()`: *"For each item, extract this value and compare by that."*

Lambda is perfect for this because the key function is almost always a single expression.

### Real-World Scenarios

- A ride-hailing app shows drivers **sorted by distance** from your location
- An e-commerce site lets you sort products **by price, lowest to highest**
- A leaderboard shows players **sorted by score, highest first**
- A flight booking app sorts results **by departure time**

````python
# Sorting a list of products by price (ascending)
products = [
    {"name": "Monitor",    "price": 22000},
    {"name": "Keyboard",   "price": 3500},
    {"name": "Laptop",     "price": 85000},
    {"name": "Mouse",      "price": 1200},
    {"name": "Headphones", "price": 4500},
]

# Sort cheapest to most expensive
by_price_asc  = sorted(products, key=lambda p: p["price"])

# Sort most expensive to cheapest (reverse=True)
by_price_desc = sorted(products, key=lambda p: p["price"], reverse=True)

print("Cheapest first:")
for p in by_price_asc:
    print(f"  {p['name']:15}  Rs. {p['price']:,}")

print("\nMost expensive first:")
for p in by_price_desc:
    print(f"  {p['name']:15}  Rs. {p['price']:,}")
````

````python
# Sorting by multiple criteria
# First sort by department name (A-Z), then within each department
# sort by salary (highest first). This is done with a tuple as the key.

employees = [
    {"name": "Priya",   "dept": "Engineering", "salary": 90000},
    {"name": "Arjun",   "dept": "Marketing",   "salary": 70000},
    {"name": "Divya",   "dept": "Engineering", "salary": 95000},
    {"name": "Kiran",   "dept": "Marketing",   "salary": 80000},
    {"name": "Meera",   "dept": "Engineering", "salary": 85000},
]

# Negative salary so highest salary sorts first within a department
sorted_employees = sorted(employees, key=lambda e: (e["dept"], -e["salary"]))

print("Employees sorted by department, then salary (highest first):")
for e in sorted_employees:
    print(f"  {e['dept']:15}  {e['name']:10}  Rs. {e['salary']:,}")
````

````python
# Finding the top and bottom performers using min() and max()
# Both functions also accept a key= argument

students = [
    {"name": "Aarav",   "score": 88},
    {"name": "Bhavna",  "score": 95},
    {"name": "Chirag",  "score": 72},
    {"name": "Deepika", "score": 91},
]

topper = max(students, key=lambda s: s["score"])
needs_help = min(students, key=lambda s: s["score"])

print(f"Class topper     : {topper['name']} ({topper['score']} marks)")
print(f"Needs extra help : {needs_help['name']} ({needs_help['score']} marks)")
````

## 9. Lambda with Conditional Logic

### Ternary Expressions inside Lambda

Even though a lambda cannot contain `if-else` statements (blocks with `:` and indentation), it can contain a **ternary expression** — Python's one-line conditional.

The syntax is:
```
value_if_true   if   condition   else   value_if_false
```

This makes lambdas very useful for classification tasks — assigning a category or label based on a value.

### Real-World Use Cases

- Labeling a transaction as "Credit" or "Debit" based on its amount
- Classifying a product as "Affordable", "Mid-range", or "Premium"
- Assigning a grade (A, B, C, F) based on exam score
- Marking an order as "Shipped" or "Pending" based on a status code

````python
# Classify exam scores into grades
# This is a nested ternary — read it left to right:
# If >= 90 -> A, elif >= 80 -> B, elif >= 70 -> C, else -> F

grade = lambda score: "A" if score >= 90 else ("B" if score >= 80 else ("C" if score >= 70 else "F"))

scores = [95, 82, 73, 61, 90, 78]

print("Score  Grade")
print("-" * 15)
for s in scores:
    print(f"  {s}     {grade(s)}")
````

````python
# Classify products into pricing tiers
products = [
    {"name": "Pen",      "price": 20},
    {"name": "Bag",      "price": 1500},
    {"name": "Laptop",   "price": 85000},
    {"name": "Notebook", "price": 120},
    {"name": "Phone",    "price": 35000},
]

tier = lambda p: "Budget" if p < 500 else ("Mid-range" if p < 20000 else "Premium")

print(f"{'Product':12}  {'Price':>8}  Tier")
print("-" * 35)
for p in products:
    print(f"{p['name']:12}  Rs. {p['price']:>5,}  {tier(p['price'])}")
````

## 10. Higher-Order Functions — Returning Lambda from a Function

### What is a Higher-Order Function?

A function that **takes another function as an argument** or **returns a function** is called a higher-order function. In Python, functions are "first-class objects" — you can pass them around like any other value.

Lambda functions are often used here to create **function factories** — functions that generate and return customized functions on demand.

### Real-World Use Case

Imagine a discount engine in an e-commerce app. Different user segments get different discounts: new users get 10%, premium members get 25%, sale day gets 50%. Instead of writing three separate functions, you build one factory that generates the right discount function for any rate.

````python
# Function factory: generates a discount function for any rate
def make_discount(rate):
    return lambda price: round(price * (1 - rate), 2)

# Create discount functions for different user segments
new_user_discount     = make_discount(0.10)   # 10% off
premium_discount      = make_discount(0.25)   # 25% off
flash_sale_discount   = make_discount(0.50)   # 50% off

price = 4000

print(f"Original price     : Rs. {price}")
print(f"New user (10% off) : Rs. {new_user_discount(price)}")
print(f"Premium (25% off)  : Rs. {premium_discount(price)}")
print(f"Flash sale (50% off): Rs. {flash_sale_discount(price)}")
````

## 11. Common Mistakes to Avoid

### Mistake 1: Assigning a Lambda to a Variable and Reusing It Everywhere

If you find yourself writing `my_func = lambda x: ...` and using `my_func` in many places across your code, stop and use `def` instead. Assigned lambdas offer no advantage over `def`, and they make the code harder to debug (no function name in error tracebacks).

````python
# Avoid this — there is no benefit over def
calculate_tax = lambda amount, rate: round(amount * rate, 2)

# Prefer this — clearer, testable, shows up properly in tracebacks
def calculate_tax(amount, rate):
    return round(amount * rate, 2)

print(calculate_tax(5000, 0.18))  # 900.0
````

### Mistake 2: The Loop Variable Trap (Closure Bug)

When you create lambdas inside a loop, all of them capture the **same variable reference**, not the value at the time of creation. By the time you call any of them, the loop has finished and the variable holds its final value.

````python
# Wrong — all lambdas see the final value of i (which is 4)
wrong_funcs = [lambda x: x + i for i in range(5)]
print([f(0) for f in wrong_funcs])  # [4, 4, 4, 4, 4] -- NOT what you expect!

# Correct — capture the current value of i using a default argument
correct_funcs = [lambda x, i=i: x + i for i in range(5)]
print([f(0) for f in correct_funcs])  # [0, 1, 2, 3, 4] -- correct!
````

### Mistake 3: Trying to Put Statements Inside a Lambda

A lambda body must be a **single expression** — a piece of code that evaluates to a value. Statements (like `print()`, `if-else blocks`, `for` loops, `return`) are not expressions and cannot appear inside a lambda.

````python
# This will raise a SyntaxError — you cannot use statements in a lambda
# bad = lambda x: if x > 0: return x else: return -x

# Correct approach for conditional logic: use a ternary expression
abs_val = lambda x: x if x >= 0 else -x

print(abs_val(-7))   # 7
print(abs_val(4))    # 4

# If the logic is complex, always switch to def
def safe_divide(a, b):
    if b == 0:
        return None
    return a / b

print(safe_divide(10, 2))   # 5.0
print(safe_divide(10, 0))   # None
````

## 12. Best Practices

Here is a concise guide to using lambda functions the right way:

### Use lambda when:

- You need a short function **inline** as an argument (e.g., in `map`, `filter`, `sorted`)
- The logic fits **comfortably in one expression** without tricks
- The function is **used only once** in a specific context and not reused

### Use `def` when:

- The function will be **called in multiple places**
- The logic **cannot be expressed in one line** cleanly
- You want to **document** the function with a docstring
- You need **error handling** inside the function
- **Readability matters** (which it always does in team environments)

### A Note on Readability

Python's own style guide (PEP 8) recommends against assigning a lambda to a variable:

> "Always use a def statement instead of an assignment statement that binds a lambda expression directly to an identifier."

This does not mean lambdas are bad — it means they are designed to be **anonymous and temporary**, passed directly as arguments rather than stored as named functions.

````python
# Summary — Good vs Bad Lambda Usage

data = [3, -1, 4, -1, 5, -9, 2, 6]

# GOOD: lambda as an inline argument — concise, clear purpose
positives = list(filter(lambda x: x > 0, data))
squared   = list(map(lambda x: x ** 2, data))

print("Positive numbers:", positives)
print("Squared values  :", squared)

# GOOD: lambda as a sort key — short and readable
words = ["banana", "apple", "cherry", "date"]
sorted_by_length = sorted(words, key=lambda w: len(w))
print("Sorted by length:", sorted_by_length)

# AVOID: lambda assigned to a variable and reused (just use def)
# validate = lambda x: x > 0 and x < 100

# PREFER:
def is_valid_percentage(x):
    """Check if x is a valid percentage (0 to 100)."""
    return 0 < x < 100

print("Is 85 valid?", is_valid_percentage(85))
print("Is 110 valid?", is_valid_percentage(110))
````

## Summary

| Concept | What it does | Best used for |
|---------|-------------|---------------|
| `lambda args: expr` | Creates a small anonymous function | Inline arguments to map, filter, sorted |
| `map(fn, list)` | Applies `fn` to every item | Transforming all items in a collection |
| `filter(fn, list)` | Keeps items where `fn` returns True | Selecting a subset based on a condition |
| `reduce(fn, list)` | Folds the list into one value | Summing, multiplying, merging |
| `sorted(list, key=fn)` | Sorts by a derived value | Ordering dicts, tuples, objects |

### The Mental Model

Whenever you see `map`, `filter`, or `reduce`, ask yourself:

- **"Do I want to change the shape of every item?"** → `map()`
- **"Do I want to keep only some items?"** → `filter()`
- **"Do I want to boil everything down to one answer?"** → `reduce()`

Lambda is simply the tool you use to express the *"what"* in each of those operations — concisely and inline.
