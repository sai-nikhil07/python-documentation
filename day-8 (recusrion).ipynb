## 1. What is Recursion?

Recursion is a programming technique where **a function solves a problem by calling itself** on a smaller version of the same problem. This continues until the problem becomes small enough to be solved directly, at which point the recursion stops.

### A simple real-world analogy: The Queue

Imagine you are standing in a queue and you want to know your position, but you cannot see the front. You ask the person in front of you: *"What position are you?"*

They do not know either, so they ask the person in front of them. This chain continues until the person at the very front says: *"I am at position 1."*

Then, everyone adds `1` and passes the answer back. When the answer finally reaches you, you know your position.

Each person solved the same problem by asking someone else to solve a slightly simpler version of it.

### The mathematical idea

Consider the factorial of a number (`n!`):
`factorial(5) = 5 * factorial(4)`

We can define factorial recursively:
```
factorial(5) = 5 * factorial(4)
factorial(4) = 4 * factorial(3)
factorial(3) = 3 * factorial(2)
factorial(2) = 2 * factorial(1)
factorial(1) = 1   <-- this is where we stop (base case)
```

## 2. How the Call Stack Works

Every time a function is called, Python creates a **stack frame** in memory to store its local variables, parameters, and return point. These frames are stored on the **call stack**.

During recursion, each call creates a **new, separate frame**. Once the base case is reached, the frames are solved and removed one by one, unwinding the results back up.

```
Call stack while computing factorial(4):

  [ factorial(1) ]  <-- top of stack, base case reached
  [ factorial(2) ]  <-- waiting for factorial(1)
  [ factorial(3) ]  <-- waiting for factorial(2)
  [ factorial(4) ]  <-- waiting for factorial(3)
```

````python
# Tracing the call stack in action
def factorial_traced(n, depth=0):
    indent = "    " * depth
    print(f"{indent}factorial({n}) called")

    if n == 1:
        print(f"{indent}Base case reached: returning 1")
        return 1

    result = n * factorial_traced(n - 1, depth + 1)
    print(f"{indent}factorial({n}) returning {result}")
    return result

print("Tracing factorial(4):")
answer = factorial_traced(4)
print(f"\nFinal answer: {answer}")
````

## 3. Anatomy of a Recursive Function

Every correct recursive function must have two key parts:

1. **The Base Case**: The condition under which the function stops calling itself and returns a direct answer immediately.
2. **The Recursive Case**: The part where the function calls itself with a **smaller or simpler input**, moving closer to the base case.

````python
# Example 1: Factorial
def factorial(n):
    # 1. Base case
    if n == 0 or n == 1:
        return 1

    # 2. Recursive case
    return n * factorial(n - 1)

print("factorial(5) =", factorial(5))
````

````python
# Example 2: Sum of numbers from 1 to n
def sum_to_n(n):
    # 1. Base case
    if n <= 1:
        return n

    # 2. Recursive case
    return n + sum_to_n(n - 1)

print("sum_to_n(10) =", sum_to_n(10))
````

## 4. Direct Recursion

Direct recursion occurs when **a function calls itself directly** within its own body. This is the most common form of recursion.

````python
# Example 1: Power calculation (base ^ exp)
def power(base, exp):
    if exp == 0:
        return 1
    return base * power(base, exp - 1)

# Example 2: Sum of digits of a number
def sum_of_digits(n):
    if n == 0:
        return 0
    return (n % 10) + sum_of_digits(n // 10)

# Example 3: Simple countdown
def countdown(n):
    if n <= 0:
        print("Done!")
        return
    print(n)
    countdown(n - 1)

print("power(2, 5) =", power(2, 5))
print("sum_of_digits(1234) =", sum_of_digits(1234))
print("Countdown:")
countdown(3)
````

## 5. Indirect Recursion

Indirect recursion occurs when **a function does not call itself directly**, but calls another function which eventually calls the first function back, forming a cycle.

````python
# Example: Check even/odd by alternating calls
def is_even(n):
    if n == 0:
        return True
    return is_odd(n - 1)  # calls is_odd

def is_odd(n):
    if n == 0:
        return False
    return is_even(n - 1) # calls is_even back

print("4 is even:", is_even(4))
print("3 is even:", is_even(3))
````

## 6. Head Recursion

In head recursion, the **recursive call is made first**, and any other operations or processing happen *after* the recursive call returns (during unwinding).

````python
# Example 1: Print numbers 1 to n
def head_print(n):
    if n == 0:
        return
    head_print(n - 1)  # 1. Call first
    print(n)           # 2. Work after
print("head_print(3) output:")
head_print(3)
````

````python
# Example 2: Reversing a string
def reverse_string(s):
    if len(s) == 0:
        return ""
    return reverse_string(s[1:]) + s[0]


print("\nreverse_string('abc') =", reverse_string("abc"))
````

## 7. Tail Recursion

In tail recursion, the **recursive call is the very last operation** performed by the function. There is no pending work left to do after the call returns.

To make a function tail recursive, we often use an **accumulator** to pass the intermediate result forward.

````python
def show(n):
    if n == 0:
        return

    print(n)      # work first
    show(n - 1)   # recursive call last

show(5)
````

````python
# Example 1: Tail-recursive Factorial
def factorial_tail(n, acc=1):
    if n <= 1:
        return acc
    return factorial_tail(n - 1, n * acc)  # recursive call is the last thing
print("factorial_tail(5) =", factorial_tail(5))

````

````python
# Example 2: Tail-recursive Sum of a List
def sum_list_tail(lst, acc=0):
    if not lst:
        return acc
    return sum_list_tail(lst[1:], acc + lst[0])


print("sum_list_tail([1, 2, 3]) =", sum_list_tail([1, 2, 3]))
````

## 8. Recursion vs Loops

Recursion and loops (iteration) are both ways to repeat actions. Conceptionally:
- **Loops** maintain state using explicit variables inside a loop block.
- **Recursion** maintains state implicitly via stack frames on the call stack.

| Aspect | Recursion | Loops (Iteration) |
|---|---|---|
| Code Style | Declarative (what to do) | Imperative (how to do) |
| State Management | Implicit (call stack) | Explicit (variables) |
| Best For | Nested / branched data (trees, graphs) | Flat sequences, counting |
| Overhead | High (requires stack frames) | Low (no frame overhead) |

````python
# Comparing tree traversal: Recursion vs Iterative
class Node:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right

# Constructing a simple tree:
#     2
#    / \
#   1   3
tree = Node(2, Node(1), Node(3))

# 1. Clean recursive traversal
def inorder_recursive(node, result=None):
    if result is None: result = []
    if node is None: return result
    inorder_recursive(node.left, result)
    result.append(node.value)
    inorder_recursive(node.right, result)
    return result

# 2. Verbose iterative traversal using an explicit stack
def inorder_iterative(root):
    result, stack = [], []
    current = root
    while current or stack:
        while current:
            stack.append(current)
            current = current.left
        current = stack.pop()
        result.append(current.value)
        current = current.right
    return result

print("Recursive Traversal:", inorder_recursive(tree))
print("Iterative Traversal:", inorder_iterative(tree))
````

## 9. When to Use Recursion

- **Good Use Cases**: Hierarchical data (directories, XML/JSON), divide-and-conquer algorithms (Merge Sort), backtracking (puzzles/maze routing).
- **Poor Use Cases**: Iterating over flat lists, sequential counts, performance-critical loops.

````python
# Example 1: Flattening nested lists
def flatten(nested):
    result = []
    for item in nested:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

# Example 2: Finding a key in a nested dictionary
def find_key(d, target_key):
    if target_key in d:
        return d[target_key]
    for key, value in d.items():
        if isinstance(value, dict):
            result = find_key(value, target_key)
            if result is not None:
                return result
    return None

print("Flattened:", flatten([1, [2, [3, 4]], 5]))
print("Found Database Host:", find_key({"app": {"db": {"host": "localhost"}}}, "host"))
````

## 10. Classic Problems

````python
# Classic Problem 1: Tower of Hanoi
def hanoi(n, source, target, helper):
    if n == 1:
        print(f"  Move disk 1 from {source} to {target}")
        return
    hanoi(n - 1, source, helper, target)
    print(f"  Move disk {n} from {source} to {target}")
    hanoi(n - 1, helper, target, source)

print("Hanoi (3 Disks):")
hanoi(3, 'A', 'C', 'B')
````

````python
# Classic Problem 2: Binary Search
def binary_search(arr, target, low, high):
    if low > high:
        return -1
    mid = (low + high) // 2
    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search(arr, target, mid + 1, high)
    else:
        return binary_search(arr, target, low, mid - 1)

data = [2, 5, 8, 12, 16, 23, 38]
print("Index of 12:", binary_search(data, 12, 0, len(data) - 1))
````

````python
# Classic Problem 3: Merge Sort
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

print("Merge Sorted:", merge_sort([38, 27, 43, 3, 9]))
````

````python
# Classic Problem 4: All Permutations of a String
def permutations(s):
    if len(s) <= 1:
        return [s]
    result = []
    for i in range(len(s)):
        current = s[i]
        remaining = s[:i] + s[i+1:]
        for perm in permutations(remaining):
            result.append(current + perm)
    return result

print("Permutations of 'ABC':", permutations("ABC"))
````

````python
# Classic Problem 5: Check Palindrome
def is_palindrome(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return is_palindrome(s[1:-1])

print("'racecar' is palindrome:", is_palindrome("racecar"))
print("'hello' is palindrome:", is_palindrome("hello"))
````

````python
# Classic Problem 6: Greatest Common Divisor (Euclidean Algorithm)
def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)

print("GCD of 48 and 18:", gcd(48, 18))
````

## 11. Common Mistakes

### Mistake 1: Missing base case

````python
# WRONG (No base case):
# def bad_factorial(n):
#     return n * bad_factorial(n - 1)

# CORRECT:
def good_factorial(n):
    if n <= 1:
        return 1
    return n * good_factorial(n - 1)

print(good_factorial(5))
````

### Mistake 2: Not moving toward base case

````python
# WRONG (n doesn't decrease):
# def bad_countdown(n):
#     if n == 0: return
#     bad_countdown(n)

# CORRECT:
def good_countdown(n):
    if n <= 0:
        print("Done!")
        return
    good_countdown(n - 1)

good_countdown(3)
````

### Mistake 3: Using a mutable default argument

Default arguments in Python are evaluated once when the function is defined. Sharing a mutable list across calls leaks data.

````python
# WRONG (shared list leaks intermediate state):
def bad_collect(n, result=[]):
    if n == 0: return result
    result.append(n)
    return bad_collect(n - 1, result)

# CORRECT (fresh list created on top-level call):
def good_collect(n, result=None):
    if result is None: result = []
    if n == 0: return result
    result.append(n)
    return good_collect(n - 1, result)

print("bad_collect first call:", bad_collect(2))
print("bad_collect second call (leaked):", bad_collect(2))
print("good_collect first call:", good_collect(2))
print("good_collect second call:", good_collect(2))
````

### Mistake 4: Wrong or narrow base case condition

````python
# WRONG (crashes on negative inputs):
# def bad_sum(n):
#     if n == 0: return 0
#     return n + bad_sum(n - 1)

# CORRECT:
def good_sum(n):
    if n <= 0: return 0
    return n + good_sum(n - 1)

print("good_sum(5) =", good_sum(5))
print("good_sum(-3) =", good_sum(-3))
````

## 12. Summary

### Core Types of Recursion
- **Direct**: The function calls itself directly.
- **Indirect**: The function cycles through intermediate calls (A -> B -> A).
- **Head**: The recursive call happens before any processing; work is done on the way back up.
- **Tail**: The recursive call is the very last operation; results are compiled on the way down.

### The Golden Rules of Recursion
1. **Always define a base case.**
2. **Ensure each recursive step moves closer to the base case.**
3. **Use recursion for naturally branching/hierarchical structures**; use standard loops for flat, linear processing.
4. **Avoid mutable default arguments** inside recursive helper functions.
