## 1. Introduction to OOP

### What is OOP?
Object-Oriented Programming (OOP) is a programming style that organises code around **objects** — which combine data (attributes) and behaviour (methods) — instead of just writing functions and procedures.

### Why OOP is Needed
| Problem without OOP | How OOP solves it |
|---|---|
| Data and functions scattered globally | Grouped together inside a class |
| Hard to reuse code | Inheritance reuses existing classes |
| Difficult to manage large codebases | Modular, self-contained objects |
| No real-world mapping | Objects model real-world entities |

### Procedural vs OOP
| Aspect | Procedural | OOP |
|---|---|---|
| Focus | Functions and steps | Objects and interactions |
| Data | Separate from functions | Bundled inside class |
| Reuse | Copy functions | Inheritance |
| Examples | C, Pascal | Python, Java, C++ |

### Real-world OOP Examples
- **Car** : attributes = color, brand; methods = start(), brake()
- **Bank Account** : attributes = owner, balance; methods = deposit(), withdraw()
- **Student** : attributes = name, marks; methods = get_grade(), is_pass()

````python
# Procedural style - data and functions are separate
student_name  = "Alice"
student_marks = 88

def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    else:
        return "C"

print(get_grade(student_marks))
````

````python
# OOP style - data and behaviour bundled inside a class
class Student:
    def __init__(self, name, marks):
        self.name  = name
        self.marks = marks

    def get_grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        else:
            return "C"

s = Student("Alice", 88)
print(s.get_grade())
````

## 2. Classes and Objects

| Term | Analogy | Meaning |
|---|---|---|
| **Class** | Blueprint of a house | Template defining attributes and methods |
| **Object** | The actual house | A real instance created from the class |

- Define a class with `class ClassName:`
- Create an object with `obj = ClassName()`
- Access an attribute with `obj.attribute`
- Call a method with `obj.method()`

````python
# Creating a class
class Student:
    pass

print(type(Student))   # <class 'type'>
````

````python
# Creating a single object from a class
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Alice")
print(s1.name)
print(type(s1))
````

````python
# Creating multiple objects from the same class
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age  = age

s1 = Student("Alice", 20)
s2 = Student("Bob",   22)
s3 = Student("Clara", 21)

print(s1.name, s1.age)
print(s2.name, s2.age)
print(s3.name, s3.age)
````

````python
# Accessing attributes and calling methods
class Student:
    def __init__(self, name, marks):
        self.name  = name
        self.marks = marks

    def introduce(self):
        print("My name is", self.name, "and I scored", self.marks)

    def result(self):
        if self.marks >= 50:
            print(self.name, "Passed")
        else:
            print(self.name, "Failed")

s = Student("Alice", 78)

# Accessing attributes
print(s.name)
print(s.marks)

# Calling methods
s.introduce()
s.result()
````

## 3. Attributes (Variables)

### Instance Variables
- Defined using `self.variable` inside `__init__`
- **Unique to every object** — each object holds its own copy

### Class Variables
- Defined directly inside the class body, outside any method
- **Shared by all objects** of that class

### Key Difference
| | Instance Variable | Class Variable |
|---|---|---|
| Syntax | `self.name = "Sai"` | `college = "CDU"` |
| Unique per object | Yes | No, shared |
| Access | `obj.name` | `ClassName.var` or `obj.var` |
| Use case | Name, age, marks | College, company, counter |

````python
# Instance variables - each object has its own copy
class Student:
    def __init__(self, name, marks):
        self.name  = name
        self.marks = marks

s1 = Student("Alice", 90)
s2 = Student("Bob",   75)

print(s1.name, s1.marks)   # Alice 90
print(s2.name, s2.marks)   # Bob 75

# Changing one does not affect the other
s1.marks = 95
print(s1.marks)   # 95
print(s2.marks)   # 75 - unchanged
````

````python
# Class variable - shared across all objects
class Student:
    college = "CDU"   # class variable

    def __init__(self, name):
        self.name = name

s1 = Student("Alice")
s2 = Student("Bob")

print(s1.college)      # CDU
print(s2.college)      # CDU
print(Student.college) # CDU - accessed via class name
````

````python
# Class variable as a counter - counts total objects created
class Student:
    total = 0   # shared counter

    def __init__(self, name):
        self.name = name
        Student.total += 1

s1 = Student("Alice")
s2 = Student("Bob")
s3 = Student("Clara")

print("Total students:", Student.total)   # 3
````

````python
# What happens when you assign a class variable on an instance
class Student:
    college = "CDU"

    def __init__(self, name):
        self.name = name

s1 = Student("Alice")
s2 = Student("Bob")

# This creates a NEW instance variable on s1 only
# It does NOT change the class variable
s1.college = "MIT"

print(s1.college)      # MIT  - s1's own instance variable
print(s2.college)      # CDU  - unchanged
print(Student.college) # CDU  - class variable unchanged
````

## 4. Methods

| Type | Decorator | First Parameter | Can Access | Typical Use |
|---|---|---|---|---|
| **Instance Method** | none | `self` | instance + class data | Most operations on object data |
| **Class Method** | `@classmethod` | `cls` | class data only | Counters, factory/alternate constructors |
| **Static Method** | `@staticmethod` | none (neither self nor cls) | neither | Utility helpers related to the class |

````python
# Instance method - operates on the object using self
class Student:
    def __init__(self, name, marks):
        self.name  = name
        self.marks = marks

    def show(self):
        print("Name:", self.name, "| Marks:", self.marks)

    def grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        else:
            return "C"

s = Student("Alice", 85)
s.show()
print(s.grade())
````

````python
# Class method - operates on the class using cls
class Student:
    college = "CDU"
    count   = 0

    def __init__(self, name):
        self.name = name
        Student.count += 1

    @classmethod
    def get_college(cls):
        return cls.college

    @classmethod
    def get_count(cls):
        print("Total students:", cls.count)

s1 = Student("Alice")
s2 = Student("Bob")

# Call via class name (no object needed)
print(Student.get_college())
Student.get_count()

# Can also call via object
print(s1.get_college())
````

````python
# Static method - utility function, no self or cls
class Student:
    @staticmethod
    def is_passing(marks):
        return marks >= 50

    @staticmethod
    def letter_grade(marks):
        if marks >= 90: return "A"
        if marks >= 75: return "B"
        if marks >= 50: return "C"
        return "F"

# No object needed
print(Student.is_passing(65))   # True
print(Student.is_passing(40))   # False
print(Student.letter_grade(88)) # B
````

````python
# Comparing all three types together
class Employee:
    company = "TechCorp"
    count   = 0

    def __init__(self, name, salary):
        self.name   = name
        self.salary = salary
        Employee.count += 1

    def show(self):               # instance method
        print(self.name, "-", self.salary)

    @classmethod
    def get_count(cls):           # class method
        print("Employees:", cls.count)

    @staticmethod
    def is_valid_salary(s):       # static method
        return s > 0

e1 = Employee("Alice", 60000)
e2 = Employee("Bob",   75000)

e1.show()
Employee.get_count()
print(Employee.is_valid_salary(50000))
print(Employee.is_valid_salary(-100))
````

## 5. Constructor - `__init__()`

- A special method that **runs automatically** when an object is created
- Purpose: set up the object's initial state (assign attributes)
- Name is always `__init__`

| Type | Parameters | Description |
|---|---|---|
| **Default** constructor | `self` only | Sets fixed/default values |
| **Parameterized** constructor | `self` + arguments | Values passed at creation time |

````python
# Default constructor - no parameters except self
class Dog:
    def __init__(self):
        self.name  = "Unknown"
        self.breed = "Mixed"

    def show(self):
        print(self.name, "-", self.breed)

d = Dog()
d.show()
````

````python
# Parameterized constructor - accepts values at creation
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year  = year

    def info(self):
        print(self.year, self.brand, self.model)

c1 = Car("Toyota", "Corolla", 2022)
c2 = Car("Honda",  "Civic",   2023)
c3 = Car("BMW",    "X5",      2024)

c1.info()
c2.info()
c3.info()
````

````python
# Constructor with default argument values
class Circle:
    def __init__(self, radius=1):
        self.radius = radius

    def area(self):
        return round(3.14159 * self.radius ** 2, 2)

c1 = Circle()      # uses default radius = 1
c2 = Circle(5)     # radius = 5
c3 = Circle(10)    # radius = 10

print(c1.area())
print(c2.area())
print(c3.area())
````

## 6. self Keyword

- `self` is a reference to **the current object**
- It is the first parameter in every instance method
- When you call `obj.method()`, Python automatically passes `obj` as `self`
- Without `self`, a method has no way to access the object's own data
- The name `self` is a convention, not a keyword, but always follow it

```
dog.bark()
# Python internally calls: Dog.bark(dog)
#                                      ^ this is self
```

````python
# self allows the method to read the object's own data
class Person:
    def __init__(self, name, age):
        self.name = name   # stored on this specific object
        self.age  = age

    def greet(self):
        print("Hello, my name is", self.name)
        print("I am", self.age, "years old")

p1 = Person("Alice", 25)
p2 = Person("Bob",   30)

p1.greet()   # self = p1
p2.greet()   # self = p2
````

````python
# self allows the method to modify the object's own data
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age  = age

    def birthday(self):
        self.age += 1
        print(self.name, "is now", self.age, "years old")

p1 = Person("Alice", 25)
p2 = Person("Bob",   30)

p1.birthday()   # only p1's age changes
p1.birthday()

print(p1.age)   # 27
print(p2.age)   # 30 - unchanged
````

````python
# self allows a method to accept another object as argument
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age  = age

    def compare_age(self, other):
        if self.age > other.age:
            print(self.name, "is older")
        elif self.age < other.age:
            print(other.name, "is older")
        else:
            print("Same age")

p1 = Person("Alice", 25)
p2 = Person("Bob",   30)

p1.compare_age(p2)
````

## 7. Object Lifecycle and Destructor

Every object goes through three stages:

| Stage | When | Method |
|---|---|---|
| **Creation** | `obj = ClassName()` called | `__init__()` runs |
| **Usage** | Accessing attributes, calling methods | Normal operations |
| **Destruction** | `del obj` or goes out of scope | `__del__()` runs |

### Destructor - `__del__()`
- Automatically called when the object is destroyed or garbage collected
- Used to release resources: close files, database connections, network sockets
- `del obj` triggers destruction manually

````python
# Basic object lifecycle
class Student:
    def __init__(self, name):
        self.name = name
        print("Object created:", self.name)

    def study(self):
        print(self.name, "is studying")

    def __del__(self):
        print("Object destroyed:", self.name)

# Stage 1: Creation
s = Student("Alice")

# Stage 2: Usage
s.study()

# Stage 3: Destruction
del s
print("After del")
````

````python
# Destructor used to close a resource
class FileHandler:
    def __init__(self, filename):
        self.filename = filename
        print("Opening file:", self.filename)

    def write(self, text):
        print("Writing to", self.filename, ":", text)

    def __del__(self):
        print("Closing file:", self.filename)

f = FileHandler("data.txt")
f.write("Hello World")
f.write("Second line")
del f
print("File handler is gone")
````

````python
# Objects go out of scope automatically when a function returns
class Connection:
    def __init__(self, name):
        self.name = name
        print("Connected:", self.name)

    def __del__(self):
        print("Disconnected:", self.name)

def use_connection():
    conn = Connection("database")
    print("Using connection...")
    # conn is destroyed when function ends

use_connection()
print("Function returned")
````

## 8. Special / Magic / Dunder Methods

Methods with double underscores that Python calls **automatically** for built-in operations.

| Method | Triggered by | Purpose |
|---|---|---|
| `__str__()` | `print(obj)`, `str(obj)` | Human-readable string |
| `__repr__()` | `repr(obj)`, REPL display | Developer/debug string |
| `__len__()` | `len(obj)` | Return a length |
| `__add__()` | `obj1 + obj2` | Addition |
| `__sub__()` | `obj1 - obj2` | Subtraction |
| `__mul__()` | `obj1 * obj2` | Multiplication |
| `__eq__()` | `obj1 == obj2` | Equality |
| `__lt__()` | `obj1 < obj2` | Less than |
| `__gt__()` | `obj1 > obj2` | Greater than |
| `__getitem__()` | `obj[key]` | Index / get |
| `__setitem__()` | `obj[key] = val` | Index / set |
| `__call__()` | `obj()` | Make object callable |

````python
# __str__ and __repr__
class Book:
    def __init__(self, title, author, pages):
        self.title  = title
        self.author = author
        self.pages  = pages

    def __str__(self):
        # Human-readable - used by print()
        return self.title + " by " + self.author

    def __repr__(self):
        # Developer-readable - used by repr() and in REPL
        return "Book(title=" + self.title + ", author=" + self.author + ", pages=" + str(self.pages) + ")"

b = Book("1984", "Orwell", 328)

print(b)        # calls __str__
print(str(b))   # calls __str__
print(repr(b))  # calls __repr__
````

````python
# __len__
class Playlist:
    def __init__(self, name):
        self.name  = name
        self.songs = []

    def add(self, song):
        self.songs.append(song)

    def __len__(self):
        return len(self.songs)

    def __str__(self):
        return "Playlist: " + self.name + " (" + str(len(self)) + " songs)"

p = Playlist("Morning Mix")
p.add("Song A")
p.add("Song B")
p.add("Song C")

print(len(p))   # 3 - calls __len__
print(p)
````

````python
# __add__, __sub__, __mul__ - arithmetic operators
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return "Vector(" + str(self.x) + ", " + str(self.y) + ")"

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

v1 = Vector(3, 4)
v2 = Vector(1, 2)

print(v1 + v2)   # Vector(4, 6)
print(v1 - v2)   # Vector(2, 2)
print(v1 * 3)    # Vector(9, 12)
````

````python
# __eq__, __lt__, __gt__ - comparison operators
class Student:
    def __init__(self, name, marks):
        self.name  = name
        self.marks = marks

    def __str__(self):
        return self.name + "(" + str(self.marks) + ")"

    def __eq__(self, other):
        return self.marks == other.marks

    def __lt__(self, other):
        return self.marks < other.marks

    def __gt__(self, other):
        return self.marks > other.marks

s1 = Student("Alice", 90)
s2 = Student("Bob",   75)
s3 = Student("Clara", 90)

print(s1 == s3)   # True  - same marks
print(s1 == s2)   # False
print(s2 < s1)    # True  - Bob has less marks
print(s1 > s2)    # True

# Sorting works automatically once __lt__ is defined
students = [s1, s2, s3]
students.sort()
for s in students:
    print(s)
````

````python
# __getitem__ and __setitem__ - index access
class Inventory:
    def __init__(self):
        self.items = {}

    def __setitem__(self, key, value):
        self.items[key] = value

    def __getitem__(self, key):
        return self.items.get(key, "Not found")

    def __len__(self):
        return len(self.items)

    def __str__(self):
        return str(self.items)

inv = Inventory()
inv["apples"]  = 50   # calls __setitem__
inv["bananas"] = 30
inv["mangoes"] = 20

print(inv["apples"])    # 50  - calls __getitem__
print(inv["grapes"])    # Not found
print(len(inv))         # 3
print(inv)
````

````python
# __call__ - makes an object callable like a function
class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, value):
        return value * self.factor

double = Multiplier(2)
triple = Multiplier(3)

print(double(5))    # 10 - object called like a function
print(triple(5))    # 15
print(double(100))  # 200
````

## 9. Encapsulation

Bundling data and methods together, and **restricting direct access** to some attributes.

| Level | Syntax | Convention | Where Accessible |
|---|---|---|---|
| **Public** | `name` | Normal | Anywhere |
| **Protected** | `_name` | Single underscore | Inside class and subclasses (convention only) |
| **Private** | `__name` | Double underscore | Only inside the class (name-mangled by Python) |

### Getters and Setters
Use `@property` for a Pythonic way to read/write private attributes with validation.

````python
# Public, protected, and private members
class Employee:
    def __init__(self, name, department, salary):
        self.name        = name          # public
        self._department = department    # protected
        self.__salary    = salary        # private

emp = Employee("Alice", "Engineering", 60000)

# Public - accessible freely
print(emp.name)

# Protected - accessible but the underscore signals "internal use"
print(emp._department)

# Private - name is mangled, direct access raises AttributeError
try:
    print(emp.__salary)
except AttributeError as e:
    print("Error:", e)

# Python mangles __salary to _Employee__salary
print(emp._Employee__salary)  # this works but is bad practice
````

````python
# Getters and setters - traditional approach
class BankAccount:
    def __init__(self, owner, balance):
        self.owner    = owner
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Balance cannot be negative")

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Deposit must be positive")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid withdrawal amount")

acc = BankAccount("Alice", 1000)
print(acc.get_balance())

acc.deposit(500)
print(acc.get_balance())

acc.withdraw(200)
print(acc.get_balance())

acc.set_balance(-100)   # rejected
````

````python
# @property - Pythonic getter and setter
class Student:
    def __init__(self, name, marks):
        self.name    = name
        self.__marks = marks

    @property
    def marks(self):              # getter - accessed like an attribute
        return self.__marks

    @marks.setter
    def marks(self, value):       # setter - called on assignment
        if 0 <= value <= 100:
            self.__marks = value
        else:
            print("Marks must be between 0 and 100")

s = Student("Alice", 85)

print(s.marks)       # calls getter - no parentheses needed

s.marks = 92         # calls setter
print(s.marks)

s.marks = 110        # rejected by setter
print(s.marks)       # still 92
````

## 10. Inheritance

A child class inherits the attributes and methods of a parent class.

| Type | Structure | Syntax |
|---|---|---|
| **Single** | One parent, one child | `class B(A):` |
| **Multilevel** | Chain: A -> B -> C | `class C(B):` |
| **Multiple** | Two parents, one child | `class C(A, B):` |
| **Hierarchical** | One parent, many children | `class B(A): class C(A):` |
| **Hybrid** | Mix of the above types | Combination |

````python
# Single Inheritance - one parent, one child
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(self.name, "is eating")

    def sleep(self):
        print(self.name, "is sleeping")

class Dog(Animal):
    def bark(self):
        print(self.name, "says: Woof!")

d = Dog("Rex")
d.eat()     # inherited from Animal
d.sleep()   # inherited from Animal
d.bark()    # Dog's own method

print(isinstance(d, Dog))     # True
print(isinstance(d, Animal))  # True - Dog IS-A Animal
````

````python
# Multilevel Inheritance - A -> B -> C
class Vehicle:
    def start(self):
        print("Vehicle started")

class Car(Vehicle):
    def drive(self):
        print("Car is driving")

class ElectricCar(Car):
    def charge(self):
        print("Charging battery")

ec = ElectricCar()
ec.start()    # from Vehicle (grandparent)
ec.drive()    # from Car (parent)
ec.charge()   # own method
````

````python
# Multiple Inheritance - one child, two parents
class Flyable:
    def fly(self):
        print("Flying in the air")

class Swimmable:
    def swim(self):
        print("Swimming in water")

class Duck(Flyable, Swimmable):
    def quack(self):
        print("Quack!")

duck = Duck()
duck.fly()    # from Flyable
duck.swim()   # from Swimmable
duck.quack()  # own method
````

````python
# Hierarchical Inheritance - one parent, many children
class Shape:
    def __init__(self, color):
        self.color = color

    def describe(self):
        print("Color:", self.color)

class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)
        self.radius = radius

    def area(self):
        return round(3.14 * self.radius ** 2, 2)

class Rectangle(Shape):
    def __init__(self, color, width, height):
        super().__init__(color)
        self.width  = width
        self.height = height

    def area(self):
        return self.width * self.height

class Triangle(Shape):
    def __init__(self, color, base, height):
        super().__init__(color)
        self.base   = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

c = Circle("Red", 5)
r = Rectangle("Blue", 4, 6)
t = Triangle("Green", 3, 8)

c.describe()
print("Circle area:", c.area())

r.describe()
print("Rectangle area:", r.area())

t.describe()
print("Triangle area:", t.area())
````

````python
# Hybrid Inheritance - mix of multiple and multilevel
class A:
    def method_a(self):
        print("Method from A")

class B(A):
    def method_b(self):
        print("Method from B")

class C(A):
    def method_c(self):
        print("Method from C")

class D(B, C):   # D inherits B and C, both inherit A
    def method_d(self):
        print("Method from D")

d = D()
d.method_a()   # from A (via B)
d.method_b()   # from B
d.method_c()   # from C
d.method_d()   # own

# MRO shows the resolution order
print(D.__mro__)
````

## 11. Method Overriding

When a **child class provides its own version** of a method that already exists in the parent.

- Same method name, different implementation
- The child's version replaces the parent's version for that class
- The parent's method is not deleted; it is only overridden for the child
- A child that does not override a method simply uses the parent's version

````python
# Basic method overriding
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(self.name, "makes a generic sound")

class Dog(Animal):
    def speak(self):              # overrides Animal.speak()
        print(self.name, "says: Woof!")

class Cat(Animal):
    def speak(self):              # overrides Animal.speak()
        print(self.name, "says: Meow!")

class Snake(Animal):
    def speak(self):
        print(self.name, "says: Hiss!")

class Fish(Animal):
    pass                          # does NOT override - uses parent version

animals = [Dog("Rex"), Cat("Whiskers"), Snake("Nagini"), Fish("Nemo")]

for a in animals:
    a.speak()
````

````python
# Overriding and extending parent behaviour
class Employee:
    def __init__(self, name, salary):
        self.name   = name
        self.salary = salary

    def get_info(self):
        print("Name:", self.name)
        print("Salary:", self.salary)

class Manager(Employee):
    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)
        self.team_size = team_size

    def get_info(self):           # overrides and extends
        super().get_info()        # call parent version first
        print("Team Size:", self.team_size)

class Intern(Employee):
    def __init__(self, name, salary, duration):
        super().__init__(name, salary)
        self.duration = duration

    def get_info(self):
        super().get_info()
        print("Internship Duration:", self.duration, "months")

m = Manager("Alice", 90000, 8)
i = Intern("Bob",    15000, 6)

print("--- Manager ---")
m.get_info()

print("--- Intern ---")
i.get_info()
````

## 12. super()

`super()` gives access to the **parent class** from inside the child class.

Common uses:
1. Call the parent's `__init__` to avoid duplicating initialisation code
2. Call the parent's version of an overridden method
3. Works correctly in multilevel and multiple inheritance (follows MRO)

````python
# Using super() to call parent __init__
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age  = age
        print("Person __init__ called")

class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)   # calls Person.__init__
        self.student_id = student_id
        print("Student __init__ called")

s = Student("Alice", 20, "S101")
print(s.name)
print(s.age)
print(s.student_id)
````

````python
# Using super() to call parent method
class Animal:
    def describe(self):
        print("I am an animal")

class Dog(Animal):
    def describe(self):
        super().describe()       # calls Animal.describe() first
        print("I am a Dog")

class GuideDog(Dog):
    def describe(self):
        super().describe()       # calls Dog.describe() (which calls Animal.describe())
        print("I am a Guide Dog")

g = GuideDog()
g.describe()
````

````python
# super() in multilevel inheritance chain
class A:
    def __init__(self):
        print("A init")

class B(A):
    def __init__(self):
        super().__init__()   # calls A
        print("B init")

class C(B):
    def __init__(self):
        super().__init__()   # calls B (which calls A)
        print("C init")

c = C()
# Output shows the chain: A -> B -> C
````

````python
# Practical example: super() in a real class hierarchy
class Vehicle:
    def __init__(self, brand, year):
        self.brand = brand
        self.year  = year

    def info(self):
        print("Brand:", self.brand, "| Year:", self.year)

class Car(Vehicle):
    def __init__(self, brand, year, doors):
        super().__init__(brand, year)
        self.doors = doors

    def info(self):
        super().info()
        print("Doors:", self.doors)

class ElectricCar(Car):
    def __init__(self, brand, year, doors, range_km):
        super().__init__(brand, year, doors)
        self.range_km = range_km

    def info(self):
        super().info()
        print("Range:", self.range_km, "km")

ec = ElectricCar("Tesla", 2024, 4, 500)
ec.info()
````

## 13. Polymorphism

**"Many Forms"** - the same interface behaves differently depending on the object.

| Type | How it works |
|---|---|
| **Method Overriding** | Subclasses give different implementations of the same method |
| **Operator Overloading** | Dunder methods redefine operators for custom objects |
| **Duck Typing** | Any object with the required method works, regardless of its type |

````python
# Polymorphism via method overriding
class Shape:
    def area(self):
        return 0

    def describe(self):
        print(self.__class__.__name__, "- area:", self.area())

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return round(3.14 * self.radius ** 2, 2)

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width  = width
        self.height = height

    def area(self):
        return self.width * self.height

class Triangle(Shape):
    def __init__(self, base, height):
        self.base   = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

shapes = [Circle(5), Rectangle(4, 6), Triangle(3, 8)]

for shape in shapes:
    shape.describe()   # same call, different result for each type
````

````python
# Polymorphism via operator overloading
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __str__(self):
        return "Rs." + str(self.amount)

    def __add__(self, other):
        return Money(self.amount + other.amount)

    def __sub__(self, other):
        return Money(self.amount - other.amount)

    def __mul__(self, factor):
        return Money(self.amount * factor)

    def __eq__(self, other):
        return self.amount == other.amount

    def __gt__(self, other):
        return self.amount > other.amount

m1 = Money(1000)
m2 = Money(500)

print(m1 + m2)   # Rs.1500
print(m1 - m2)   # Rs.500
print(m1 * 2)    # Rs.2000
print(m1 > m2)   # True
print(m1 == m2)  # False
````

````python
# Polymorphism via duck typing
# "If it has the right method, it works" - no shared parent needed

class Dog:
    def speak(self):
        return "Woof"

class Cat:
    def speak(self):
        return "Meow"

class Robot:
    def speak(self):
        return "Beep boop"

class Car:
    def speak(self):
        return "Honk honk"

# This function works with ANY object that has a speak() method
def make_speak(obj):
    print(obj.__class__.__name__, "says:", obj.speak())

things = [Dog(), Cat(), Robot(), Car()]

for thing in things:
    make_speak(thing)
````

## 14. Abstraction

**Hiding complex implementation details** and showing only the essential interface.

- Use `ABC` (Abstract Base Class) from the `abc` module
- An **abstract class** cannot be instantiated directly
- An **abstract method** uses `@abstractmethod` and has no body
- Every subclass **must** implement all abstract methods, or it also becomes abstract
- Non-abstract methods in the ABC are inherited normally by all subclasses

```python
from abc import ABC, abstractmethod
```

````python
# Abstract class cannot be instantiated directly
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

# Trying to create an object of the abstract class
try:
    s = Shape()
except TypeError as e:
    print("Error:", e)
````

````python
# Subclass must implement all abstract methods
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

# This subclass does NOT implement all abstract methods
class Incomplete(Shape):
    def area(self):
        return 0
    # perimeter() not implemented

try:
    obj = Incomplete()
except TypeError as e:
    print("Error:", e)
````

````python
# Correct usage - subclass implements all abstract methods
from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self, color):
        self.color = color

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

    def describe(self):   # non-abstract, inherited by all
        print(self.__class__.__name__, "| Color:", self.color, "| Area:", self.area(), "| Perimeter:", self.perimeter())

class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)
        self.radius = radius

    def area(self):
        return round(3.14 * self.radius ** 2, 2)

    def perimeter(self):
        return round(2 * 3.14 * self.radius, 2)

class Rectangle(Shape):
    def __init__(self, color, width, height):
        super().__init__(color)
        self.width  = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

shapes = [Circle("Red", 5), Rectangle("Blue", 4, 6)]

for s in shapes:
    s.describe()
````

````python
# Practical abstraction example - payment system
from abc import ABC, abstractmethod

class Payment(ABC):
    def __init__(self, amount):
        self.amount = amount

    @abstractmethod
    def pay(self):
        pass

    @abstractmethod
    def validate(self):
        pass

    def receipt(self):
        print("Receipt: Rs." + str(self.amount) + " paid via " + self.__class__.__name__)

class CreditCard(Payment):
    def __init__(self, amount, card_number):
        super().__init__(amount)
        self.card_number = card_number

    def validate(self):
        return len(str(self.card_number)) == 16

    def pay(self):
        if self.validate():
            print("Paid Rs." + str(self.amount) + " via Credit Card ending " + str(self.card_number)[-4:])
            self.receipt()
        else:
            print("Invalid card number")

class UPI(Payment):
    def __init__(self, amount, upi_id):
        super().__init__(amount)
        self.upi_id = upi_id

    def validate(self):
        return "@" in self.upi_id

    def pay(self):
        if self.validate():
            print("Paid Rs." + str(self.amount) + " via UPI: " + self.upi_id)
            self.receipt()
        else:
            print("Invalid UPI ID")

class Cash(Payment):
    def validate(self):
        return self.amount > 0

    def pay(self):
        if self.validate():
            print("Paid Rs." + str(self.amount) + " in Cash")
            self.receipt()

payments = [
    CreditCard(5000, 1234567890123456),
    UPI(200, "alice@upi"),
    Cash(50)
]

for p in payments:
    p.pay()
    print()
````
