##### Working of Built-in Functions and Their Internal Working

Built-in functions like `type()`, `id()`, `print()`, and `len()` are implemented in C (in CPython) for performance.

- **`type(obj)`**: Returns the class of the object. Internally, it accesses the `ob_type` pointer in the C structure of the object.
- **`id(obj)`**: Returns the unique identifier of an object, which in CPython is the memory address of the object.
- **`len(obj)`**: Calls the `__len__()` dunder method of the object. For built-in types like lists and strings, it directly reads the size field from the C struct, which is an O(1) operation.

**Memory Management and Interning:**
Python uses reference counting and a garbage collector. For performance, Python caches small integers (from -5 to 256) and interns certain strings (like identifiers). This means multiple variables assigned the value `10` will point to the exact same memory address.

```python
a = 10
b = 10
print('id(a):', id(a))
print('id(b):', id(b))
print('Are a and b the same object?', a is b)

c = 257
d = 257
print('Are c and d the same object? (Depends on environment, usually False in REPL)', c is d)
```

##### Lists

A list is an ordered, mutable collection of items. It can contain elements of different data types. Internally, lists are dynamic arrays. When a list grows beyond its current allocated capacity, Python over-allocates memory to allow for O(1) amortized appends.

```python
my_list = [1, 'Python', 3.14, True]
print(my_list)

# Modifying a list
my_list[0] = 100
my_list.append('New Item')
print(my_list)
```

##### Tuples

A tuple is an ordered, immutable collection. Once created, its elements cannot be modified. They are more memory efficient than lists and can be used as dictionary keys if all their elements are immutable.

```python
my_tuple = (1, 2, 3, 'Python')
print(my_tuple)

# Accessing elements
print(my_tuple[1])

# my_tuple[0] = 10 # This would raise a TypeError
```

##### Strings

A string is a sequence of characters. In Python 3, all strings are Unicode. Strings are immutable, meaning any operation that modifies a string actually creates a new string object.

```python
my_string = 'Hello, Python!'
print(my_string)
print('Length:', len(my_string))
print('Uppercase:', my_string.upper())
```

##### Frozen Sets

A `frozenset` is an immutable version of a `set`. A `set` is an unordered collection of unique elements. Because a `frozenset` is immutable, it is hashable and can be used as a dictionary key or an element of another set.

```python
normal_set = {1, 2, 3, 3}
frozen = frozenset([1, 2, 3, 4, 4])

print('Set:', normal_set)
print('Frozenset:', frozen)

# normal_set.add(5) # Valid
# frozen.add(5) # AttributeError: 'frozenset' object has no attribute 'add'
```

##### Dictionaries

A dictionary is an unordered (ordered by insertion in Python 3.7+), mutable collection of key-value pairs. Keys must be immutable and hashable (like strings, numbers, or tuples). Internally, dictionaries are implemented as highly optimized hash tables.

```python
my_dict = {'name': 'John', 'age': 25, 'language': 'Python'}
print(my_dict)

# Accessing and adding elements
print('Name:', my_dict['name'])
my_dict['city'] = 'New York'
print(my_dict)
```

##### Bytes

The `bytes` type represents an immutable sequence of bytes (integers in the range 0 <= x < 256). It is useful for binary data like images, network packets, etc.

```python
b_data = b'Hello'
print(b_data)
print('First byte:', b_data[0]) # Returns ASCII value of 'H'

# b_data[0] = 65 # TypeError: 'bytes' object does not support item assignment
```

##### Byte Arrays

The `bytearray` type is a mutable counterpart to `bytes`. It allows modification of the bytes in-place.

```python
ba_data = bytearray(b'Hello')
print(ba_data)

# Modifying a bytearray
ba_data[0] = 74 # ASCII for 'J'
print(ba_data) # Output: bytearray(b'Jello')
```
