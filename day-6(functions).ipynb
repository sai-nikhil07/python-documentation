## 1. What is a Function?

A **function** is a reusable block of code that performs a specific task.  
Instead of writing the same logic repeatedly, you define it once and **call** it whenever needed.

### Why use functions?
| Benefit | Description |
|---|---|
| **Reusability** | Write once, use many times |
| **Readability** | Breaks complex programs into smaller, named units |
| **Maintainability** | Fix or update logic in one place |
| **Testability** | Easier to test isolated units of code |
| **Abstraction** | Hide implementation details behind a name |

````python
# Without a function — repetitive code
print("Hello, Alice!")
print("Hello, Bob!")
print("Hello, Charlie!")

print("---")

# With a function — clean and reusable
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")
greet("Bob")
greet("Charlie")
````

<a id='2'></a>
## 2. Defining and Calling a Function

- **Defining** a function creates it (the code inside does NOT run yet).
- **Calling** a function executes its code block.

````python
# DEFINING a function
def say_hello():
    print("Hello, World!")

# Nothing happens yet — the function is just stored in memory

# CALLING the function
say_hello()       # Output: Hello, World!
say_hello()       # Functions can be called multiple times
say_hello()
````

## 3. Function Syntax Breakdown

```
def function_name(parameter1, parameter2, ...):
    """
    Optional docstring — describes what the function does.
    """
    # function body (indented block)
    statement1
    statement2
    return value   # optional
```

| Part | Description |
|---|---|
| `def` | Keyword that starts a function definition |
| `function_name` | Name you choose (follows variable naming rules) |
| `( )` | Parentheses — hold parameters (can be empty) |
| `:` | Colon marks the start of the function body |
| indented body | All code belonging to the function, indented by 4 spaces |
| `return` | Optional — sends a value back to the caller |

````python
# A complete, well-structured function
def add(a, b):
    """
    Returns the sum of two numbers a and b.
    """
    result = a + b
    return result

total = add(3, 5)
print(total)   # 8
````

<a id='4'></a>
## 4. Parameters vs Arguments

These two words are often confused. Here's the distinction:

| Term | Where it appears | Meaning |
|---|---|---|
| **Parameter** | Function **definition** | Placeholder variable inside `def` |
| **Argument** | Function **call** | Actual value passed when calling |

Think of **parameters as variable names** and **arguments as the actual values** you plug in.

````python
#             ↓ parameter
def square(number):
    return number ** 2

#          ↓ argument
result = square(7)
print(result)   # 49

# Another call with a different argument
print(square(10))   # 100
````

<a id='5'></a>
## 5. Positional Arguments

**Positional arguments** are matched to parameters by their **position (order)**.  
The first argument goes to the first parameter, second to second, and so on.

>  Order matters! Swapping positional arguments gives different (often wrong) results.

````python
def describe_pet(animal, name):
    print(f"I have a {animal} named {name}.")

# Correct order
describe_pet("dog", "Bruno")    # I have a dog named Bruno.

# Swapped order — gives wrong meaning
describe_pet("Bruno", "dog")    # I have a Bruno named dog.
````

````python
# Wrong number of positional arguments raises TypeError
def multiply(x, y):
    return x * y

try:
    multiply(4)   # Missing one argument
except TypeError as e:
    print(f"Error: {e}")
````

<a id='6'></a>
## 6. Keyword Arguments

**Keyword arguments** are passed using the parameter name explicitly.  
This means **order doesn't matter**, making your function calls more readable and self-documenting.

````python
def create_profile(name, age, city):
    print(f"Name: {name} | Age: {age} | City: {city}")

# Using keyword arguments — order doesn't matter
create_profile(age=25, city="Hyderabad", name="Arjun")

# Mix positional and keyword — positional must come FIRST
create_profile("Priya", city="Mumbai", age=30)
````

````python
#  SyntaxError: keyword argument before positional
try:
    create_profile(name="Ravi", 28, "Delhi")   # positional after keyword
except SyntaxError as e:
    print(f"SyntaxError: {e}")
````

<a id='7'></a>
## 7. Default Parameter Values

You can give a parameter a **default value** so it becomes optional when calling the function.  
If the caller doesn't provide that argument, the default is used.

>  **Rule:** Parameters with defaults must come **after** parameters without defaults.

````python
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("Sita")                    # Uses default greeting
greet("Sita", "Good morning")   # Overrides default
````

````python
def power(base, exponent=2):
    """Raises base to the given exponent. Default exponent is 2 (square)."""
    return base ** exponent

print(power(5))      # 25  — uses default exponent=2
print(power(5, 3))   # 125 — overrides to exponent=3
print(power(2, 10))  # 1024
````

````python
#  Common Pitfall: Mutable default arguments
# WRONG — list is shared across all calls!
def bad_append(item, lst=[]):
    lst.append(item)
    return lst

print(bad_append(1))   # [1]
print(bad_append(2))   # [1, 2]  ← unexpected!
print(bad_append(3))   # [1, 2, 3] ← still accumulating!

print("")

# CORRECT — use None as default, create new list inside
def good_append(item, lst=None):
    if lst is None:
        lst = []
    lst.append(item)
    return lst

print(good_append(1))   # [1]
print(good_append(2))   # [2]  ← fresh list each time
print(good_append(3))   # [3]
````

<a id='8'></a>
## 8. `*args` — Variable Positional Arguments

`*args` lets a function accept **any number of positional arguments**.  
Inside the function, `args` is a **tuple** containing all extra positional values passed.

> The name `args` is a convention. The `*` is what matters — you could use `*numbers` or `*values`.

````python
def total(*args):
    """Sums any number of values."""
    print(f"args = {args}  (type: {type(args)})")
    return sum(args)

print(total(1, 2))            # 3
print(total(10, 20, 30))      # 60
print(total(5, 5, 5, 5, 5))  # 25
print(total())                # 0 — zero arguments is fine too
````

````python
# Combining a required parameter with *args
def introduce(greeting, *names):
    for name in names:
        print(f"{greeting}, {name}!")

introduce("Hi", "Alice", "Bob", "Charlie")
````

````python
# Unpacking a list/tuple into positional arguments with *
def add(a, b, c):
    return a + b + c

numbers = [1, 2, 3]
print(add(*numbers))   # Equivalent to add(1, 2, 3) → 6
````

<a id='9'></a>
## 9. `**kwargs` — Variable Keyword Arguments

`**kwargs` lets a function accept **any number of keyword arguments**.  
Inside the function, `kwargs` is a **dictionary** of `{name: value}` pairs.

> Again, `kwargs` is a convention. The `**` is what matters.

````python
def print_info(**kwargs):
    """Prints any key-value pairs passed in."""
    print(f"kwargs = {kwargs}  (type: {type(kwargs)})")
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

print_info(name="Arjun", age=22, city="Hyderabad")
````

````python
# Practical example: flexible configuration builder
def build_config(app_name, **settings):
    config = {"app": app_name}
    config.update(settings)
    return config

cfg = build_config(
    "MyApp",
    debug=True,
    version="1.0.2",
    port=8080,
    theme="dark"
)
print(cfg)
````

````python
# Unpacking a dictionary into keyword arguments with **
def display(name, age, city):
    print(f"{name} is {age} years old, from {city}.")

person = {"name": "Priya", "age": 28, "city": "Mumbai"}
display(**person)   # Equivalent to display(name='Priya', age=28, city='Mumbai')
````

<a id='10'></a>
## 10. Combining All Parameter Types

Python allows combining all parameter types in one function.  
They **must appear in this strict order**:

```
def func(positional, /, standard, *args, keyword_only, **kwargs):
```

| Order | Type | Description |
|---|---|---|
| 1 | Positional-only (`/`) | Before `/` — only positional, no keyword |
| 2 | Regular | Can be positional or keyword |
| 3 | `*args` | Collects extra positional args as tuple |
| 4 | Keyword-only | After `*args` — must be passed by name |
| 5 | `**kwargs` | Collects extra keyword args as dict |

````python
def combined(a, b, *args, keyword_only="default", **kwargs):
    print(f"a           = {a}")
    print(f"b           = {b}")
    print(f"args        = {args}")
    print(f"keyword_only= {keyword_only}")
    print(f"kwargs      = {kwargs}")

combined(1, 2, 3, 4, 5, keyword_only="custom", x=10, y=20)
````

````python
# Positional-only parameters (Python 3.8+) using /
def positional_only(x, y, /):
    """x and y can ONLY be passed positionally, not by keyword."""
    return x + y

print(positional_only(3, 4))    #  works

try:
    positional_only(x=3, y=4)   #  fails
except TypeError as e:
    print(f"TypeError: {e}")
````

````python
# Keyword-only parameters using * (bare star)
def keyword_only_demo(a, b, *, c, d=100):
    """c and d MUST be passed as keyword arguments."""
    print(a, b, c, d)

keyword_only_demo(1, 2, c=3)           #  d uses default
keyword_only_demo(1, 2, c=3, d=4)     #  both provided

try:
    keyword_only_demo(1, 2, 3)         #  c is keyword-only
except TypeError as e:
    print(f"TypeError: {e}")
````

<a id='11'></a>
## 11. The `return` Statement

The `return` statement:
- **Exits** the function immediately
- **Sends back** a value to the caller

If a function has no `return`, or just `return` (no value), it returns `None`.

````python
# Function WITH a return value
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

temp = celsius_to_fahrenheit(100)
print(f"100°C = {temp}°F")   # 100°C = 212.0°F

# Function WITHOUT a return (implicitly returns None)
def print_double(x):
    print(x * 2)

result = print_double(5)   # prints 10
print(f"Return value: {result}")   # Return value: None
````

````python
# Early return — exit the function before reaching the end
def divide(a, b):
    if b == 0:
        return None   # early exit
    return a / b

print(divide(10, 2))   # 5.0
print(divide(10, 0))   # None (no crash)
````

<a id='12'></a>
## 12. Returning Multiple Values

Python functions can return multiple values — they are **packed into a tuple** automatically.

````python
def min_max(numbers):
    """Returns both the minimum and maximum of a list."""
    return min(numbers), max(numbers)   # returns a tuple

data = [3, 1, 7, 2, 9, 4]

result = min_max(data)
print(f"Result:  {result}  (type: {type(result)})")   # tuple

# Tuple unpacking (most common pattern)
low, high = min_max(data)
print(f"Min: {low}, Max: {high}")
````

````python
# Returning a dictionary for named multi-value return
def stats(numbers):
    return {
        "count": len(numbers),
        "total": sum(numbers),
        "average": sum(numbers) / len(numbers),
        "minimum": min(numbers),
        "maximum": max(numbers),
    }

s = stats([10, 20, 30, 40, 50])
for key, value in s.items():
    print(f"  {key:10}: {value}")
````

<a id='13'></a>
## 13. Docstrings

A **docstring** is a string literal placed right after the `def` line.  
It documents what the function does, its parameters, and return value.

Access it via `function.__doc__` or the built-in `help()` function.

````python
# One-liner docstring
def square(n):
    """Returns the square of n."""
    return n ** 2

# Multi-line docstring (Google style)
def calculate_bmi(weight_kg, height_m):
    """
    Calculates Body Mass Index (BMI).

    Args:
        weight_kg (float): Weight in kilograms.
        height_m  (float): Height in metres.

    Returns:
        float: BMI rounded to 2 decimal places.

    Raises:
        ValueError: If height_m is zero.
    """
    if height_m == 0:
        raise ValueError("Height cannot be zero.")
    return round(weight_kg / (height_m ** 2), 2)

print(calculate_bmi(70, 1.75))

print("\n--- Docstring ---")
print(calculate_bmi.__doc__)
````

````python
# Using help() to see the docstring
help(calculate_bmi)
````

## 14. Type Hints / Annotations

Python allows you to **annotate** function parameters and return values with expected types.  
These are **hints only** — Python doesn't enforce them at runtime, but tools like `mypy` do.

```python
def func(param: type) -> return_type:
```

````python
# Basic type hints
def greet(name: str, times: int = 1) -> str:
    return (f"Hello, {name}!\n") * times

print(greet("World", 3))
````

````python
from typing import List, Dict, Optional, Tuple, Union

def process_scores(scores: List[float]) -> Dict[str, float]:
    """Returns a summary dict from a list of scores."""
    return {
        "average": sum(scores) / len(scores),
        "highest": max(scores),
        "lowest":  min(scores),
    }

def find_user(user_id: int) -> Optional[str]:
    """Returns a username or None if not found."""
    db = {1: "Alice", 2: "Bob"}
    return db.get(user_id)  # returns None if not found

def to_number(value: Union[int, str]) -> float:
    """Converts an int or string to float."""
    return float(value)

print(process_scores([85.0, 92.5, 78.0, 90.0]))
print(find_user(1))    # Alice
print(find_user(99))   # None
print(to_number("42"))  # 42.0
````

````python
# Inspecting annotations
def area(length: float, width: float) -> float:
    return length * width

print(area.__annotations__)
````

<a id='15'></a>
## 15. Lambda Functions

A **lambda** is a small, anonymous (nameless) function defined in a **single expression**.  
Use it when you need a quick, throwaway function — especially as an argument to another function.

```python
lambda parameters: expression
```

>  Lambda can only contain a **single expression**, not multiple statements.

````python
# Regular function vs Lambda
def double(x):
    return x * 2

double_lambda = lambda x: x * 2

print(double(5))         # 10
print(double_lambda(5))  # 10
````

````python
# Common use cases for lambda

# 1. Sorting with a custom key
students = [("Alice", 88), ("Bob", 72), ("Charlie", 95), ("Diana", 85)]
sorted_by_score = sorted(students, key=lambda s: s[1], reverse=True)
print("Sorted by score:", sorted_by_score)

# 2. map() — apply function to each element
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print("Squared:", squared)

# 3. filter() — keep elements where condition is True
evens = list(filter(lambda x: x % 2 == 0, numbers))
print("Evens:", evens)

# 4. Multiple parameters
add = lambda a, b: a + b
print("3 + 7 =", add(3, 7))
````

## 16. Nested Functions

You can define a function **inside another function**.  
The inner function has access to variables of the outer function (**closure**).

````python
def outer(x):
    def inner(y):
        return x + y   # inner accesses x from outer's scope
    return inner(10)

print(outer(5))   # 15
````

````python
# Closures — returning the inner function itself
def make_multiplier(factor):
    """Returns a function that multiplies by `factor`."""
    def multiply(number):
        return number * factor   # 'factor' is remembered via closure
    return multiply

triple = make_multiplier(3)
quadruple = make_multiplier(4)

print(triple(7))     # 21
print(quadruple(7))  # 28
print(triple(10))    # 30
````

## 17. Scope — Local vs Global

**Scope** determines where a variable is accessible.

Python uses the **LEGB rule** to resolve names:

| Letter | Scope | Description |
|---|---|---|
| **L** | Local | Inside the current function |
| **E** | Enclosing | In any outer (enclosing) function |
| **G** | Global | At the top level of the module |
| **B** | Built-in | Python's built-in names (`len`, `print`, ...) |

````python
x = "global"   # global scope

def outer():
    x = "enclosing"   # enclosing scope

    def inner():
        x = "local"   # local scope
        print("inner:", x)   # sees local

    inner()
    print("outer:", x)   # sees enclosing

outer()
print("global:", x)   # sees global
````

````python
# Variables inside a function are LOCAL — don't affect global
count = 0

def increment():
    count = 10   # creates a LOCAL variable, not the global one
    print("Inside function:", count)

increment()
print("Outside function:", count)   # still 0
````

<a id='18'></a>
## 18. The `global` and `nonlocal` Keywords

- **`global`**: lets a function read **and modify** a global variable
- **`nonlocal`**: lets an inner function modify a variable in the **enclosing** function's scope

````python
# global keyword
count = 0

def increment():
    global count
    count += 1

increment()
increment()
increment()
print("Count:", count)   # 3
````

````python
# nonlocal keyword
def make_counter():
    count = 0

    def increment():
        nonlocal count   # refers to the enclosing function's count
        count += 1
        return count

    return increment

counter = make_counter()
print(counter())   # 1
print(counter())   # 2
print(counter())   # 3
````

## 19. First-Class Functions

In Python, functions are **first-class objects** — meaning they can be:
- Assigned to variables
- Passed as arguments to other functions
- Returned from functions
- Stored in data structures

````python
# 1. Assigned to a variable
def shout(text):
    return text.upper() + "!!!"

yell = shout   # function reference (no parentheses)
print(yell("hello"))   # HELLO!!!
````

````python
# 2. Passed as an argument (Higher-order functions)
def apply(func, value):
    """Applies any function to value."""
    return func(value)

print(apply(shout, "python"))              # PYTHON!!!
print(apply(str.lower, "HELLO WORLD"))    # hello world
print(apply(len, "twelve chars"))         # 12
````

````python
# 3. Stored in data structures
def add(a, b): return a + b
def sub(a, b): return a - b
def mul(a, b): return a * b
def div(a, b): return a / b if b != 0 else None

operations = {
    "+": add,
    "-": sub,
    "*": mul,
    "/": div,
}

for op_symbol, op_func in operations.items():
    print(f"10 {op_symbol} 3 = {op_func(10, 3)}")
````

## 20. Recursion

A **recursive function** calls itself.  
Every recursive function needs:
1. A **base case** — condition where recursion stops
2. A **recursive case** — where the function calls itself with a simpler input

>  Without a base case, recursion continues forever → `RecursionError`.

````python
# Classic example: Factorial
# 5! = 5 × 4 × 3 × 2 × 1 = 120

def factorial(n):
    if n == 0 or n == 1:   # base case
        return 1
    return n * factorial(n - 1)   # recursive case

for i in range(8):
    print(f"{i}! = {factorial(i)}")
````

````python
# Fibonacci sequence: 0, 1, 1, 2, 3, 5, 8, 13, ...
def fibonacci(n):
    if n <= 0: return 0
    if n == 1: return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

fib_sequence = [fibonacci(i) for i in range(10)]
print("Fibonacci:", fib_sequence)
````

````python
# Max recursion depth demo
import sys
print(f"Default recursion limit: {sys.getrecursionlimit()}")

# Demonstrating RecursionError
def infinite_recursion():
    return infinite_recursion()   # no base case!

try:
    infinite_recursion()
except RecursionError as e:
    print(f"RecursionError: {e}")
````

## 21. Summary Cheat Sheet

```python
# ── DEFINING ──────────────────────────────────────────────────
def func_name(params):
    """Docstring."""
    body
    return value

# ── PARAMETER TYPES (in order) ────────────────────────────────
def f(pos_only, /, normal, *args, kw_only, **kwargs): ...

# ── CALLING ───────────────────────────────────────────────────
f(1, 2, 3)                 # positional
f(a=1, b=2)                # keyword
f(*my_list)                # unpack list as positional args
f(**my_dict)               # unpack dict as keyword args

# ── DEFAULTS ──────────────────────────────────────────────────
def f(a, b=10): ...        # b is optional

# ── VARIABLE ARGS ─────────────────────────────────────────────
def f(*args): ...          # tuple of positional args
def f(**kwargs): ...       # dict of keyword args

# ── RETURN ────────────────────────────────────────────────────
return x                   # single value
return x, y                # multiple (tuple)
return                     # returns None

# ── LAMBDA ────────────────────────────────────────────────────
f = lambda x, y: x + y

# ── SCOPE ─────────────────────────────────────────────────────
global var_name            # access/modify global variable
nonlocal var_name          # access/modify enclosing variable

# ── TYPE HINTS ────────────────────────────────────────────────
def f(x: int, y: str = "hi") -> bool: ...
```

### Quick Reference Table

| Concept | Syntax | Notes |
|---|---|---|
| Define | `def name(p1, p2):` | Indent body 4 spaces |
| Call | `name(arg1, arg2)` | Arguments in parentheses |
| Default | `def f(x, y=5):` | Defaults must come last |
| Positional | `f(1, 2)` | Order matters |
| Keyword | `f(y=2, x=1)` | Order doesn't matter |
| *args | `def f(*args):` | Tuple of extra positionals |
| **kwargs | `def f(**kw):` | Dict of extra keywords |
| Return | `return value` | Exits function + sends value |
| Lambda | `lambda x: x*2` | Single expression only |
| Docstring | `"""Text."""` | First line of body |
| Type hint | `def f(x: int) -> str:` | Documentation only |
| Recursion | Call self + base case | Needs termination condition |
